```python
def check_coupon_rate(rate):
    if rate < 0 or rate > 100:
        raise ValueError("할인율은 0~100 사이여야 합니다.")
    return rate

coupon_rate = 120
check_coupon_rate(coupon_rate)
```

```python
stocks = [12, 5, 8]
input_index = '5'

try:
    index = int(input_index)
    stock = stocks[index]
except ValueError:
    print("숫자만 입력하세요.")
except IndexError:
    print("해당 상품이 존재하지 않습니다.")
else:
    print(f"재고 수량: {stock}")
finally:
    print("재고 조회 완료")
```

```python
def set_stock(stock):
    if stock < 0:
        raise ValueError('재고는 음수가 될 수 없습니다.')
    print(f"재고가 {stock}개로 설정되었습니다.")

try:
    product_stock = -3
    set_stock(product_stock)
except ValueError as error:
    print(f"[재고 등록 실패] {error}")
```

```python
stock = 20
orders = [2, 3, 4]

for order in orders:
    stock = stock - order
    print(f"주문 수량: {order}, 현재 재고: {stock}")

print(f"최종 재고: {stock}")
```

```python
character_name = "전사"
character_level = 10

def show_character(name, level):
    print(f"{name} - 레벨 {level}")

show_character(character_name, character_level)

class Character:
    def __init__(self, name, level):
        self.name = name
        self.level = level

    def show(self):
        print(f"{self.name} - 레벨 {self.level}")

character = Character("전사", 10)
character.show()
```

```python
def 가공_프로세스(n):
    return n * 2 # 'Step Into' 명령 수행 시에만 이 내부 라인으로 제어권이 진입합니다.

number = 10
# 'Step Over' 명령을 수행하면 함수 내부로 들어가지 않고 최종 연산 결과값만 도출하여 다음 줄로 넘어갑니다.
result = 가공_프로세스(number) 
print(result)
```
