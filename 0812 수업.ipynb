```python
visitor_count = 10

def add_visitor():
    global visitor_count #공용 화이트보드 자체를 수정한다면 global 을 앞에 써야함
    visitor_count += 1

add_visitor()

print(visitor_count)
```

```python
def normal_delivery(product):
    return f"{product}: 일반 배송으로 출발합니다."

def fast_delivery(product):
    return f"{product}: 빠른 배송으로 출발합니다."

def deliver(delivery_fuction, product):
    return delivery_fuction(product)

result_1 = deliver(normal_delivery, "노트북")
result_2 = deliver(fast_delivery, "키보드")

print(result_1)
print(result_2)
```

```python
def greet(name):
    return f"안녕하세요, {name}님!"

def call_function(func, user_name):
    return func(user_name)

def get_greeter():
    return greet

result_1 = call_function(greet, "스파르타")
print(result_1)

my_func = get_greeter()
print(my_func("코딩"))
```

```python
#문제 10
scores = {
    "민수": 85,
    "지수": 72,
    "철수": 58,
    "영희": 91,
    "준호": 65
}

passed_students = []

for name, score in scores.items():
    if score >= 70:
        passed_students.append(name)

print(passed_students)

print(f"총 합격자는 {len(passed_students)}명입니다.")
```
