```python
user_database = [
    {"name": "민수", "level": "VIP", "active": True, "point": 1500},
    {"name": "영희", "level": "Gold", "active": False, "point": 800},
    {"name": "철수", "level": "VIP", "active": True, "point": 2300},
    {"name": "지영", "level": "Silver", "active": True, "point": 400},
    {"name": "성민", "level": "Gold", "active": True, "point": 1200}
]

print("--- 마케팅 대상 활성 VIP 및 Gold 회원 식별 결과 ---")

for user in user_database:
    if user["active"]:
        if user["level"] == "VIP":
            print(f"[우수 등급] {user['name']} 회원님에게 VIP 전용 쿠폰이 발송되었습니다.")
        elif user["level"] == "Gold":
            print(f"[일반 등급] {user['name']} 회원님에게 Gold 기본 쿠폰이 발송되었습니다.")
```

```python
total_vip_points = 0
active_user_count = 0

for user in user_database:
    if user["active"]:
        active_user_count += 1

        if user["level"] == "VIP":
            total_vip_points += user["point"]

print(f"전체 활성 회원 수: {active_user_count}명")
print(f"활성 VIP 회원의 누적 포인트 합계: {total_vip_points}점")
```

```python
student_scores = [85, 92, 78, 60, 99, 71, 88, 65]

highest_score = student_scores[0]
lowest_score = student_scores[0]
score_sum = 0

for score in student_scores:
    if score > highest_score:
        highest_score = score

    if score < lowest_score:
        lowest_score = score

    score_sum += score

total_students = len(student_scores)
score_average = score_sum / total_students

print("--- 학급 성적 종합 통계 분석 보고서 ---")
print(f"최고 성적 점수: {highest_score}점")
print(f"최저 성적 점수: {lowest_score}점")
print(f"학급 성적 총점: {score_sum}점")
print(f"학급 평균 성적: {score_average:.2f}점")
```

```python
import random

computer_number = random.randint(1, 100)
try_count = 0

print("★ 컴퓨터가 1부터 100 사이의 비밀 숫자를 생성했습니다. ★")

while True:
    user_input = input("컴퓨터의 숫자를 맞춰보세요 (1~100): ")
    guess = int(user_input)
    try_count += 1

    if guess < computer_number:
        print("Up! 더 큰 숫자입니다.")
    elif guess > computer_number:
        print("Down! 더 작은 숫자입니다.")
    else:
        print(f"♠ 정답입니다! {try_count}회 만에 숫자를 맞추셨습니다.")
        break 
```

```python
def add_all(*numbers):
    print(f"수집된 튜플 구조: {numbers}")
    total = sum(numbers)
    print(f"총합 결과: {total}")

add_all(1, 2, 3)
add_all(10, 20, 30)
```
