## 필수 2. 인덱싱과 슬라이싱으로 데이터 구간 추출하기

### 문제 2-1. 시간대별 주문 건수 일부 추출

0시부터 5시까지의 주문 건수가 순서대로 저장되어 있다고 가정합니다.

### 요구사항

1. `[12, 15, 9, 20, 18, 11]`을 `hourly_orders` 리스트에 저장하세요.
2. 첫 번째 주문 건수를 인덱싱으로 출력하세요.
3. 마지막 주문 건수를 인덱싱으로 출력하세요.
4. 인덱스 1부터 3까지의 값을 슬라이싱으로 출력하세요.
5. 처음부터 끝까지 두 칸 간격으로 추출하세요.
6. 전체 데이터를 역순으로 출력하세요.

### 예상 출력

```text
12
11
[15, 9, 20]
[12, 9, 18]
[11, 18, 20, 9, 15, 12]
```

````python
# 문제 2-1 코드를 작성하세요.
hourly_orders = [12, 15, 9, 20, 18, 11]

print(hourly_orders[0])
print(hourly_orders[-1])
print(hourly_orders[1:4])
print(hourly_orders[::2])
print(hourly_orders[::-1])
````

### 확인 사항

- `[1:4]`에서 끝 인덱스 4는 포함되지 않습니다.
- `[::2]`는 두 칸 간격 추출입니다.
- `[::-1]`은 원본을 변경하지 않고 역순의 새 리스트를 만듭니다.

## 필수 3. 시퀀스 공통 연산과 권한 확인하기

### 문제 3-1. 사용자 그룹 병합과 권한 검증

### 요구사항

1. `[101, 102]`를 `admin_group1`에 저장하세요.
2. `[103, 104]`를 `admin_group2`에 저장하세요.
3. `+` 연산자로 두 그룹을 결합하여 `all_admins`에 저장하고 출력하세요.
4. `admin_group1 * 3`의 결과를 출력하세요.
5. `"admin"`, `"editor"`, `"member"`가 포함된 `user_roles` 리스트를 생성하세요.
6. `len()`으로 역할 개수를 출력하세요.
7. `"admin" in user_roles`의 결과를 출력하세요.
8. `"guest" not in user_roles`의 결과를 출력하세요.

### 예상 출력

```text
[101, 102, 103, 104]
[101, 102, 101, 102, 101, 102]
3
True
True
```

````python
# 문제 3-1 코드를 작성하세요.
admin_group1 = [101, 102]
admin_group2 = [103, 104]
all_admins = admin_group1 + admin_group2

print(all_admins)

print(admin_group1*3)

user_roles = ["admin", "editor", "member"]
print(len(user_roles))

print("admin" in user_roles)
print("guest" not in user_roles)
````

### 확인 사항

- `+`와 `*` 연산은 원본을 직접 수정하지 않고 새로운 시퀀스를 만듭니다.
- 서로 다른 자료형인 리스트와 튜플은 `+`로 바로 결합할 수 없습니다.
- `in`, `not in`의 결과는 불리언 값입니다.

## 심화 1. 패킹과 일반 언패킹

### 문제 4-1. 주문 요약 정보 묶기와 분리하기

하나의 주문 정보를 주문번호, 상품명, 결제금액 순서로 묶고 다시 분리합니다.

### 요구사항

1. 다음 세 값을 쉼표로 나열하여 `packed_order`에 저장하세요.
   - 주문번호: `1001`
   - 상품명: `"Americano"`
   - 결제금액: `4500`
2. `packed_order`와 자료형을 출력하세요.
3. `order_id`, `menu_name`, `payment_amount` 세 변수로 일반 언패킹하세요.

## 실습 마무리 질문

1. 리스트와 튜플은 각각 어떤 데이터에 적합한가요?
2. `0`과 `-1` 인덱스는 각각 어떤 요소를 가리키나요?
3. `[1:4]`, `[::2]`, `[::-1]`은 각각 어떤 결과를 만드나요?
4. 시퀀스의 `+`, `*`, `len()`, `in`은 각각 어떤 역할을 하나요?

### 답변 작성

1. 리스트는 가변적인 데이터, 튜플은 데이터가 변하지 않는 경우 적합하다.
2. 0은 가장 첫 데이터, -1은 가장 마지막 데이터를 가리킨다.
3. [1:4]는 인덱스 1번을 포함, 4번은 포함하지 않는 1, 2, 3번 인덱스를 반환. [::2]는 전체 데이터를 2칸 간격으로 추출. [::-1]는 데이터를 역순으로 정렬한다.
4. +는 동일한 자료형을 연결, *은 해당 데이터 구조와 요소를 지정한 횟수만큼 반복하여 복제본을 누적한 새 객체를 반환. len()은 시퀀스 요소의 개수를 반환. in은 시퀀스 안에 특정 요소가 있는지의 여부를 True/False로 결과값을 반환.
