## 필수 1. 쇼핑몰 주문 데이터 필터링 및 집계

### 문제 1-1. 우수 주문 건수와 결제 금액 계산

배송이 완료된 주문 중 결제 금액이 50,000원 이상인 주문만 선별하여 건수와 합계를 계산합니다.

### 주문 데이터

```python
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]
```

### 요구사항

1. 위 데이터를 `order_database`에 저장하세요.
2. 주문 건수 누적 변수 `premium_order_count`를 `0`으로 초기화하세요.
3. 결제 금액 누적 변수 `premium_total_amount`를 `0`으로 초기화하세요.
4. `for`문으로 주문 데이터를 하나씩 순회하세요.
5. 배송 완료 여부가 `True`인지 확인하세요.
6. 배송 완료 주문 중 결제 금액이 `50000` 이상인지 확인하세요.
7. 두 조건을 만족하면 주문 건수를 1 증가시키고 결제 금액을 누적하세요.
8. 조건을 만족한 고객 이름도 함께 출력하세요.
9. 반복이 끝나면 주문 건수와 총 결제 금액을 출력하세요.

### 예상 출력

```text
우수 주문 고객: 김철수
우수 주문 고객: 최지영
우수 고객 주문 건수: 2건
우수 고객 총 결제 금액: 145000원
```

> 중첩 조건문 또는 `and`로 결합한 조건문 중 하나를 선택하여 작성할 수 있습니다.

````python
# 문제 1-1 코드를 작성하세요.
order_database = [
    {"name": "김철수", "completed": True, "price": 60000},
    {"name": "이영희", "completed": False, "price": 35000},
    {"name": "박민수", "completed": True, "price": 25000},
    {"name": "최지영", "completed": True, "price": 85000},
    {"name": "정성민", "completed": False, "price": 90000}
]

premium_order_count = 0
premium_total_amount = 0

for order in order_database:
    if order["completed"] == True:
        if order["price"] >= 50000:
            premium_order_count += 1
            premium_total_amount += order["price"]
            print(f"우수 주문 고객: {order["name"]}")

print(f"우수 고객 주문 건수: {premium_order_count}")
print(f"우수 고객 총 결제 금액: {premium_total_amount}원")
````

### 확인 사항

- 리스트 안의 각 딕셔너리에서 `"name"`, `"completed"`, `"price"`를 조회했는지 확인하세요.
- 누적 변수는 반복문 전에 초기화해야 합니다.
- 배송 미완료 주문은 결제 금액이 높아도 집계하지 않습니다.

## 필수 2. 품질 측정값 통계 직접 계산하기

### 문제 2-1. 최고·최저·총합·평균 당도 산출

외부 라이브러리와 `max()`, `min()`, `sum()`을 사용하지 않고 반복문으로 통계값을 계산합니다.

### 당도 데이터

```python
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]
```

### 요구사항

1. 위 데이터를 `fruit_sweetness`에 저장하세요.
2. `max_sweetness`와 `min_sweetness`를 첫 번째 요소로 초기화하세요.
3. `total_sweetness`를 `0`으로 초기화하세요.
4. `for`문으로 모든 값을 순회하세요.
5. 현재 값이 기존 최고값보다 크면 최고값을 갱신하세요.
6. 현재 값이 기존 최저값보다 작으면 최저값을 갱신하세요.
7. 현재 값을 총합에 누적하세요.
8. 총합을 `len(fruit_sweetness)`로 나누어 평균을 계산하세요.
9. 최고값, 최저값, 총합, 평균을 출력하세요.
10. 총합과 평균은 소수점 둘째 자리까지 출력하세요.

### 예상 출력

```text
최고 과일 당도: 15.0
최저 과일 당도: 10.5
과일 당도 총합: 128.70
과일 평균 당도: 12.87
```

> 이번 문제는 리스트에 하나 이상의 값이 존재한다는 전제로 진행합니다.

````python
# 문제 2-1 코드를 작성하세요.
fruit_sweetness = [12.5, 14.2, 11.8, 10.5, 15.0, 13.6, 11.2, 14.8, 12.0, 13.1]

max_sweetness = fruit_sweetness[0]
min_sweetness = fruit_sweetness[0]
total_sweetness = 0

for sweetness in fruit_sweetness:
    if sweetness > max_sweetness:
        max_sweetness = sweetness

    if sweetness < min_sweetness:
        min_sweetness = sweetness

    total_sweetness += sweetness

total_fruits = len(fruit_sweetness)
sweetness_average = total_sweetness / total_fruits

print(f"최고 과일 당도: {max_sweetness}")
print(f"최저 과일 당도: {min_sweetness}")
print(f"과일 당도 총합: {total_sweetness:.2f}")
print(f"과일 평균 당도: {sweetness_average:.2f}")
````

### 확인 사항

- `max()`, `min()`, `sum()`을 사용하지 않았는지 확인하세요.
- 첫 번째 값을 초기 기준으로 사용하면 모든 값이 음수이거나 소수여도 비교할 수 있습니다.
- 빈 리스트에서 `[0]`을 조회하면 오류가 발생하므로 데이터가 존재한다는 전제가 필요합니다.

## 심화 1. Up-Down 숫자 맞히기 게임

### 문제 3-1. 난수와 무한 루프를 활용한 반복형 프로그램

컴퓨터가 생성한 1부터 100 사이의 비밀 숫자를 맞힐 때까지 반복합니다.

### 요구사항

1. `random` 모듈을 불러오세요.
2. `random.randint(1, 100)`으로 비밀 숫자를 생성하여 `computer_number`에 저장하세요.
3. 시도 횟수 `try_count`를 `0`으로 초기화하세요.
4. 안내 문구를 출력하세요.

```text
컴퓨터가 1부터 100 사이의 비밀 숫자를 생성했습니다.
```

5. `while True:`로 반복을 시작하세요.
6. `input()`으로 숫자를 입력받으세요.
7. 입력값을 `int()`로 변환하여 `guess`에 저장하세요.
8. 입력할 때마다 시도 횟수를 1 증가시키세요.
9. `if-elif-else`로 입력값과 비밀 숫자를 비교하세요.
   - 입력값이 작으면: `"Up! 더 큰 숫자입니다."`
   - 입력값이 크면: `"Down! 더 작은 숫자입니다."`
   - 정답이면: `"정답입니다! X회 만에 맞혔습니다."`
10. 정답을 맞히면 `break`로 반복을 종료하세요.

````python
# 문제 3-1 코드를 작성하세요.
import random

computer_number = random.randint(1, 100)

try_count = 0

print("컴퓨터가 1부터 100사이의 비밀 숫자를 생성했습니다.")

while True:
    user_input = input("컴퓨터의 숫자를 맞춰보세요 (1~100): ")
    guess = int(user_input)
    try_count += 1

    if guess < computer_number:
        print("Up! 더 큰 숫자입니다.")

    elif guess > computer_number:
        print("Down! 더 작은 숫자입니다.")

    else:
        print(f"정답입니다! {try_count}회 만에 맞혔습니다.")
        break
    
````

### 실행 주의사항

- `input()`으로 받은 값은 문자열이므로 정수 비교 전에 `int()` 변환이 필요합니다.
- 숫자가 아닌 문자를 입력하면 `ValueError`가 발생할 수 있습니다.
- 예외 처리를 아직 배우지 않았다면 숫자만 입력하세요.
- `break`가 없으면 정답을 맞혀도 반복이 종료되지 않습니다.
