## 필수 1. 리스트를 `for`문으로 순회하기

### 문제 1-1. 주문 처리 상태 출력

여러 주문의 처리 상태를 앞에서부터 하나씩 확인합니다.

### 요구사항

1. 다음 리스트를 `order_statuses`에 저장하세요.

```python
["결제완료", "상품준비", "배송중"]
```

2. `for`문으로 각 요소를 `status` 변수에 하나씩 저장하세요.
3. 각 반복에서 다음 형식으로 출력하세요.

```text
현재 주문 상태: 결제완료
```

### 예상 출력

```text
현재 주문 상태: 결제완료
현재 주문 상태: 상품준비
현재 주문 상태: 배송중
```

````python
# 문제 1-1 코드를 작성하세요.
order_statuses = ["결제완료", "상품준비", "배송중"]

for status in order_statuses:
    print(f"현재 주문 상태: {status}")
````

## 필수 2. `range()`로 반복 횟수와 간격 제어하기

### 문제 2-1. 고객 대기번호 발행

### 요구사항

1. `range(1, 5)`를 사용하여 1번부터 4번까지 출력하세요.
2. 반복 변수는 `customer_number`로 작성하세요.
3. 다음 형식으로 출력하세요.


### 문제 2-2. 홀수 회차 점검 일정 출력

### 요구사항

1. `range(1, 7, 2)`를 사용하세요.
2. 1회차, 3회차, 5회차만 출력하세요.

```text
1회차 점검을 진행합니다.
```

```text
3회차 점검을 진행합니다.
```

## 필수 3. `while`문으로 처리 진행률 갱신하기

### 문제 3-1. 데이터 업로드 진행률

업로드 진행률이 80%에서 시작하여 100%까지 5%씩 증가한다고 가정합니다.

### 요구사항

1. `progress`에 `80`을 저장하세요.
2. `progress <= 100`인 동안 반복하세요.
3. 다음 형식으로 현재 진행률을 출력하세요.

```text
현재 진행률: 80%
```

4. 반복문 내부에서 `progress`를 5씩 증가시키세요.
5. 100% 출력 후 반복이 종료되는지 확인하세요.

````python
# 문제 3-1 코드를 작성하세요.
progress = 80

while progress <= 100:
    print(f"현재 진행률: {progress}%")
    progress += 5
````

### 확인 사항

- 반복 조건이 참인 동안만 실행되는지 확인하세요.
- 상태값 증가 코드가 없으면 무한 루프가 발생할 수 있습니다.
- 실행 전 반복 횟수가 정해진 번호 발행은 `for`, 상태가 목표에 도달할 때까지 반복하는 진행률은 `while`이 더 자연스럽습니다.

## 필수 4. `break`로 검색을 조기에 종료하기

### 문제 4-1. 특정 주문번호 검색

주문번호 목록에서 대상 주문을 찾으면 더 이상 검색하지 않고 반복을 종료합니다.

### 요구사항

1. `[1001, 1002, 1003, 1004, 1005]`를 `order_ids`에 저장하세요.
2. 찾을 주문번호 `target_order_id`에 `1003`을 저장하세요.
3. `for`문으로 주문번호를 하나씩 확인하세요.
4. 현재 주문번호가 대상 주문번호와 같으면 다음 문구를 출력하세요.

```text
대상 주문 1003을 찾았습니다.
```

5. 문구 출력 직후 `break`로 반복을 종료하세요.
6. 대상 주문을 찾기 전에는 다음 형식으로 출력하세요.

```text
주문 1001 확인 중
```

### 예상 출력

```text
주문 1001 확인 중
주문 1002 확인 중
대상 주문 1003을 찾았습니다.
```

> 1004와 1005는 출력되지 않아야 합니다.

````python
# 문제 4-1 코드를 작성하세요.
order_ids = [1001, 1002, 1003, 1004, 1005]
target_order_id = 1003

for target_order_id in order_ids:
    if target_order_id != 1003:
        print(f"주문 {target_order_id} 확인 중")
    elif target_order_id == 1003:
        print(f"대상 주문 {target_order_id}을 찾았습니다.")
        break
````

## 심화 1. `continue`로 처리할 수 없는 값 건너뛰기

### 문제 5-1. 결측 점수 제외

점수 목록에서 값이 없는 항목은 건너뛰고 정상 점수만 출력합니다.

### 요구사항

1. `[80, None, 90, None, 75]`를 `scores`에 저장하세요.
2. `for`문으로 각 점수를 순회하세요.
3. 점수가 `None`이면 `continue`를 사용하여 아래 출력을 건너뛰세요.
4. 정상 점수만 다음 형식으로 출력하세요.

```text
처리 점수: 80
```

### 예상 출력

```text
처리 점수: 80
처리 점수: 90
처리 점수: 75
```

````python
# 문제 5-1 코드를 작성하세요.
scores = [80, None, 90, None, 75]

for score in scores:
    if score == None:
        continue
    print(f"처리 점수: {score}")
````

### 확인 사항

- `continue` 아래의 코드는 해당 회차에서 실행되지 않습니다.
- `break`는 전체 반복을 종료하지만 `continue`는 현재 회차만 건너뜁니다.

## 심화 2. 중첩 반복문으로 좌석 배치 출력하기

### 문제 6-1. 교육장 좌석 안내

2개 행에 각각 3개의 좌석이 있다고 가정합니다.

### 요구사항

1. 전체 좌석 수를 저장할 `seat_count`를 `0`으로 초기화하세요.
2. 바깥쪽 반복문에서 `row`를 1부터 2까지 반복하세요.
3. 안쪽 반복문에서 `seat`를 1부터 3까지 반복하세요.
4. 각 좌석을 다음 형식으로 출력하세요.

```text
1행 1번 좌석
```

5. 좌석을 출력할 때마다 `seat_count`를 1씩 증가시키세요.
6. 반복이 끝난 뒤 전체 좌석 수를 출력하세요.

### 예상 출력

```text
1행 1번 좌석
1행 2번 좌석
1행 3번 좌석
2행 1번 좌석
2행 2번 좌석
2행 3번 좌석
전체 좌석 수: 6
```

````python
# 문제 6-1 코드를 작성하세요.
seat_count = 0

for row in range(1, 3):
    for seat in range(1, 4):
        print(f"{row}행 {seat}번 좌석")
        seat_count += 1

print(f"전체 좌석 수: {seat_count}")
````

### 확인 사항

- 바깥쪽 반복문 2회 × 안쪽 반복문 3회로 총 6회 실행됩니다.
- 가장 안쪽 코드는 두 단계의 들여쓰기를 적용해야 합니다.
