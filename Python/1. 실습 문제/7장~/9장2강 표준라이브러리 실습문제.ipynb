## 필수 1. `datetime`과 `time`을 활용한 시각 기록 및 실행 시간 측정

### 문제 1-1. 업무 처리 로그 만들기

서비스 작업의 실행 시각과 간단한 처리 소요 시간을 기록합니다.

### 요구사항
1. `datetime`과 `time` 모듈을 불러오세요.
2. `datetime.datetime.now()`로 현재 날짜와 시각을 가져오세요.
3. 다음 형식으로 출력하세요.

```text
2026년 08월 08일 13시 25분 07초
```

4. 형식 코드는 다음을 사용하세요.

```python
"%Y년 %m월 %d일 %H시 %M분 %S초"
```

5. `time.perf_counter()`로 시작 시각을 기록하세요.
6. `time.sleep(0.5)`로 약 0.5초 대기하세요.
7. 다시 `time.perf_counter()`를 호출하여 종료 시각을 기록하세요.
8. 두 값의 차이로 소요 시간을 계산하세요.
9. 소요 시간을 소수점 둘째 자리까지 출력하세요.

````python
# datetime과 time을 사용해 현재 시각과 실행 소요 시간을 확인하세요.
import datetime
import time

now = datetime.datetime.now()
print(f"현재 날짜: {now.strftime('%Y년 %m월 %d일 %H시 %M분 %S초')}")

start_time = time.perf_counter()
time.sleep(0.5)
end_time = time.perf_counter()

elapsed = end_time - start_time
print(f"소요 시간: {elapsed:.2f}초")

````

### 확인 사항
- 현재 시각은 실행 시점에 따라 달라집니다.
- `time.sleep(0.5)`는 약 0.5초 동안 실행을 멈춥니다.
- 실제 측정값은 시스템 상황에 따라 정확히 `0.50`초가 아닐 수 있습니다.

## 필수 2. `math`를 활용한 수치 처리

### 문제 2-1. 배송 수량과 거리 계산 보조 기능

### 데이터

```python
box_count = 12.7
temperature = -5.5
area = 81
```

### 요구사항
1. `math` 모듈을 불러오세요.
2. `math.ceil(box_count)`로 올림 값을 구하세요.
3. `math.floor(box_count)`로 내림 값을 구하세요.
4. `math.trunc(box_count)`로 소수점 이하를 버린 값을 구하세요.
5. `math.fabs(temperature)`로 절대값을 구하세요.
6. `math.sqrt(area)`로 제곱근을 구하세요.
7. `math.pi` 값을 출력하세요.

````python
# math 모듈의 주요 수치 처리 기능을 사용하세요.
import math

box_count = 12.7
temperature = -5.5
area = 81

print(f"올림: {math.ceil(box_count)}")
print(f"내림: {math.floor(box_count)}")
print(f"버림: {math.trunc(box_count)}")
print(f"절대값: {math.fabs(temperature)}")
print(f"제곱근: {math.sqrt(area)}")
print(math.pi)
````

## 필수 3. `json` 직렬화와 역직렬화

### 문제 3-1. 학생 정보를 JSON 문자열로 변환하고 복원하기

### 데이터

```python
student_dict = {
    "이름": "홍길동",
    "나이": 21,
    "전공": "컴퓨터공학과",
    "취미": ["개인 학습", "영화감상"]
}
```

### 요구사항
1. `json` 모듈을 불러오세요.
2. `json.dumps()`를 사용해 딕셔너리를 JSON 문자열로 변환하세요.
3. 한글이 유지되도록 `ensure_ascii=False`를 사용하세요.
4. 읽기 쉽도록 `indent=4`를 사용하세요.
5. 변환된 JSON 문자열과 자료형을 출력하세요.
6. `json.loads()`를 사용해 JSON 문자열을 다시 파이썬 객체로 복원하세요.
7. 복원한 객체의 자료형을 출력하세요.
8. 복원된 `"이름"` 값을 출력하세요.

````python
# JSON 직렬화와 역직렬화를 수행하세요.
import json

student_dict = {
    "이름": "홍길동",
    "나이": 21,
    "전공": "컴퓨터공학과",
    "취미": ["개인 학습", "영화감상"]
}

json_text = json.dumps(student_dict, ensure_ascii = False, indent = 4)

print("JSON 데이터:")
print(json_text)
print("변환 후 타입:", type(json_text))

restored = json.loads(json_text)

print("복원 후 타입", type(restored))
print("복원된 이름:", restored["이름"])
````

### 확인 사항

직렬화 전후 자료형 흐름은 다음과 같습니다.

```text
dict → str → dict
```

`json.dumps()`는 JSON 문자열을 만들고, `json.loads()`는 JSON 문자열을 다시 파이썬 객체로 복원합니다.

## 필수 4. `random`을 활용한 무작위 데이터 생성

### 문제 4-1. 이벤트 추첨 기능 구현

### 데이터

```python
participants = ["김철수", "이영희", "박민수", "최지우", "정하늘"]
```

### 요구사항
1. `random` 모듈을 불러오세요.
2. `random.choice()`로 참가자 한 명을 무작위 선택하세요.
3. `random.randint(1, 100)`으로 1~100 사이 행운 점수를 생성하세요.
4. `random.random()`으로 0 이상 1 미만의 실수 난수를 생성하세요.
5. `random.sample(participants, 2)`로 **중복 없이** 당첨자 2명을 추출하세요.
6. 각 결과를 출력하세요.

````python
# random의 choice(), randint(), random(), sample()을 사용하세요.
import random

participants = ['김철수', '이영희', '박민수', '최지우', '정하늘']

print(random.choice(participants))
print("행운 점수:", random.randint(1, 100))
print(random.random())
print("당첨자:", random.sample(participants, 2))
````

### 확인 사항
- `randint(1, 100)`은 1과 100을 모두 포함합니다.
- `random()`은 0 이상 1 미만의 실수를 반환합니다.
- `sample()`은 원본 목록을 변경하지 않고 중복 없이 지정 개수만큼 추출합니다.

### 문제 6-1. 작업 경로와 Python 버전 확인

### 요구사항
1. `os`, `sys` 모듈을 불러오세요.
2. `os.getcwd()`로 현재 작업 디렉토리를 확인하세요.
3. `os.path.join()`을 사용해 현재 작업 경로와 `"data"` 폴더명을 결합하세요.
4. `os.path.exists()`로 해당 경로가 실제로 존재하는지 확인하세요.
5. `sys.version`으로 현재 Python 버전 정보를 출력하세요.

````python
# os와 sys를 사용해 현재 실행 환경 정보를 확인하세요.
import os
import sys

current_folder = os.getcwd()
file_path = os.path.join(current_folder, "data")

print("현재 작업 폴더:", current_folder)
print("확인할 파일 경로:", file_path)

if os.path.exists(file_path):
    print("data 파일이 존재합니다.")
else:
    print("data 파일이 존재하지 않습니다.")
````
