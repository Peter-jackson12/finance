## 필수 1. 텍스트 파일 저장과 읽기 함수 비교

### 문제 1-1. 메모 파일 생성하기

### 요구사항
1. `memo.txt` 파일을 쓰기 모드(`"w"`)로 여세요.
2. `encoding="utf-8"`을 지정하세요.
3. 다음 세 문장을 각각 한 줄씩 저장하세요.

```text
오늘은 파일 입출력을 공부했습니다.
텍스트 파일을 직접 저장해 보았습니다.
내일은 구조화된 데이터를 복습하겠습니다.
```

4. 파일 저장 후 `read()`를 사용해 전체 내용을 출력하세요.

````python
# memo.txt 파일을 생성하고 read()로 전체 내용을 출력하세요.
with open("memo.txt", "w", encoding = 'utf-8') as file:
    file.write("오늘은 파일 입출력을 공부했습니다.\n")
    file.write("텍스트 파일을 직접 저장해 보았습니다.\n")
    file.write("내일은 구조화된 데이터를 복습하겠습니다.\n")

with open("memo.txt", "r", encoding = 'utf-8') as file:
    content = file.read()
    print("read():")
    print(content)
````

### 문제 1-2. `readline()`과 `readlines()` 비교

같은 `memo.txt` 파일을 사용합니다.

### 요구사항
1. 파일을 다시 열고 `readline()`을 한 번 호출하세요.
2. 반환값과 자료형을 출력하세요.
3. 파일을 다시 열고 `readlines()`를 호출하세요.
4. 반환값과 자료형을 출력하세요.
5. 세 함수의 반환 형태 차이를 정리하세요.

### 정리
- `read()`:
- `readline()`:
- `readlines()`:

````python
# read(), readline(), readlines()의 차이를 확인하세요.
with open('memo.txt', 'r', encoding= 'utf-8') as file:
    first_line = file.readline()
    print("readline():")
    print(first_line.strip())

with open('memo.txt', 'r', encoding = 'utf-8') as file:
    lines = file.readlines()
    print("readlines():")
    print(lines)
````

### 확인 사항

- `readline()` → 한 줄을 문자열로 반환
- `readlines()` → 각 줄을 요소로 가지는 리스트로 반환

## 필수 2. 텍스트 파일을 읽어 가공한 뒤 새 파일로 저장

### 문제 2-1. 원본 로그에 줄 번호 붙이기

### 요구사항
1. `source.txt` 파일에 다음 세 줄을 저장하세요.

```text
서버 시작
데이터 로드 완료
분석 작업 종료
```

2. `readlines()`로 모든 줄을 읽으세요.
3. `enumerate(lines, start=1)`를 사용하여 줄 번호를 부여하세요.
4. 가공한 내용을 `result.txt`에 다음 형식으로 저장하세요.

```text
1번 줄: 서버 시작
2번 줄: 데이터 로드 완료
3번 줄: 분석 작업 종료
```

5. `result.txt`를 다시 읽어 전체 내용을 출력하세요.

````python
# source.txt를 생성하고 각 줄에 번호를 붙여 result.txt로 저장하세요.
with open('source.txt', 'w', encoding = 'utf-8') as file:
    file.write("서버시작 \n")
    file.write("데이터 로드 완료 \n")
    file.write("분석 작업 종료 \n")

with open('source.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()

with open('result.txt', 'w', encoding='utf-8') as file:
    for number, line in enumerate(lines, start =1 ):
        result = f'{number}번 줄: {line}'
        file.write(result)


````

### 확인 사항
이 문제는 실무의 기본 파일 처리 흐름을 단순화한 형태입니다.

```text
원본 파일 읽기 → 데이터 가공 → 새 파일 저장 → 결과 검증
```

## 필수 3. CSV 파일 저장 및 복원

### 문제 3-1. 회원 점수 데이터를 CSV로 저장하기

### 데이터

```python
member_data = [
    ["id", "name", "score"],
    ["user1", "철수", 85],
    ["user2", "영희", 90],
    ["user3", "민수", 78]
]
```

### 요구사항
1. `csv` 모듈을 불러오세요.
2. `members.csv`를 쓰기 모드로 여세요.
3. `encoding="utf-8"`과 `newline=""`을 지정하세요.
4. `csv.writer()`를 생성하세요.
5. `writerows()`로 전체 2차원 리스트를 저장하세요.
6. 파일을 다시 읽기 모드로 여세요.
7. `csv.reader()` 결과를 `list()`로 변환하세요.
8. 복원된 2차원 리스트를 출력하세요.

````python
# member_data를 members.csv에 저장하고 다시 읽어오세요.
import csv

member_data = [
    ["id", "name", "score"],
    ["user1", "철수", 85],
    ["user2", "영희", 90],
    ["user3", "민수", 78]
]

with open('member.csv', 'w', encoding='utf-8', newline="") as file:
    writer = csv.writer(file)
    writer.writerow(member_data)

with open('member.csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    parsed_csv = list(reader)

print(parsed_csv)
````

### 예상 형태

```text

 ['user2', '영희', '90'],
 ['user3', '민수', '78']]
```

## 심화 1. JSON 파일 직렬화 및 역직렬화

### 문제 5-1. 강의 정보 JSON 파일 저장

### 데이터

```python
course_data = {
    "course_name": "AX Agent Course",
    "active": True,
    "details": {
        "hours": 8,
        "location": "인천"
    }
}
```

### 요구사항
1. `json` 모듈을 불러오세요.
2. `course.json`을 쓰기 모드로 여세요.
3. `json.dump()`로 딕셔너리를 파일에 저장하세요.
4. 한글을 읽기 쉽게 저장하기 위해 `ensure_ascii=False`를 사용하세요.
5. 파일을 보기 쉽게 `indent=4`를 지정하세요.
6. 같은 파일을 다시 읽기 모드로 여세요.
7. `json.load()`로 파이썬 객체로 복원하세요.
8. 전체 복원 데이터와 자료형을 출력하세요.
9. 다음 값과 자료형을 각각 확인하세요.
   - `parsed_json["active"]`
   - `parsed_json["details"]["hours"]`

````python
# course_data를 JSON 파일로 저장하고 다시 복원하세요.
import json

course_data = {
    "course_name": "AX Agent Course",
    "active": True,
    "details": {
        "hours": 8,
        "location": "인천"
    }
}

with open('course.json', 'w', encoding='utf-8') as file:
    json.dump(course_data, file, ensure_ascii= False, indent=4)

with open('course.json', 'r', encoding = 'utf-8') as file:
    parsed_json = json.load(file)

print(parsed_json)
print(type(parsed_json))
print(parsed_json['active'], type(parsed_json['active']))
print(parsed_json['details']['hours'], type(parsed_json['details']['hours']))
````

### 확인 사항
JSON 파일 안에서는 Python의 `True`가 JSON 표준의 `true`로 기록됩니다.  
`json.load()`로 읽으면 다시 Python의 `True`로 복원됩니다.

JSON은 CSV와 달리 딕셔너리·리스트·숫자·불리언과 같은 구조를 비교적 자연스럽게 복원할 수 있습니다.
