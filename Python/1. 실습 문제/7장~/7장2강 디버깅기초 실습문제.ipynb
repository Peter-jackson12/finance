## 심화 3. 논리 오류 디버깅

### 문제 5-1. 고객 주문 금액 평균 계산 오류

다음 코드는 오류 없이 실행되지만 평균 주문 금액이 잘못 계산됩니다.

### 실습 코드

````python
def calculate_order_average(order_amounts):
    total = 0

    for amount in order_amounts:
        total = amount

    average = total / len(order_amounts)
    return average

orders = [20000, 30000, 40000]
result = calculate_order_average(orders)
print(f"평균 주문 금액: {result}")
````
