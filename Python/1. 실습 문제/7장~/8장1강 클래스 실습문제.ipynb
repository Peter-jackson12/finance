### 문제 1-1. `Book` 클래스 구현

## 필수 2. 은행 계좌 클래스와 상태 변경

### 문제 2-1. `BankAccount` 클래스 구현

예금주와 잔액을 하나의 객체로 관리하고, 입금과 출금에 따라 잔액 상태가 변경되도록 구현합니다.

### 요구사항

1. `BankAccount` 클래스를 정의하세요.
2. `__init__(self, owner, balance)`를 작성하세요.
3. `owner`, `balance`를 인스턴스 속성으로 저장하세요.
4. `deposit(self, amount)` 메서드를 정의하세요.
5. 입금액만큼 `balance`를 증가시키고 다음 형식으로 출력하세요.

```text
[김철수 계좌] 2000원 입금 / 현재 잔액: 7000원
```

6. `withdraw(self, amount)` 메서드를 정의하세요.
7. 잔액이 출금액 이상이면 출금액만큼 잔액을 감소시키세요.
8. 잔액이 부족하면 잔액을 변경하지 않고 다음 문구를 출력하세요.

```text
[김철수 계좌] 출금 실패: 잔액이 부족합니다.
```

9. `김철수 / 5000원`, `이영희 / 10000원` 계좌를 각각 생성하세요.
10. 철수 계좌에 2000원을 입금하고, 영희 계좌에서 3000원을 출금하세요.
11. 두 계좌의 최종 잔액을 각각 출력하세요.

````python
# BankAccount 클래스를 작성하세요.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f'[{self.owner} 계좌] {amount}원 입금 / 현재 잔액: {self.balance}원')
        return self.balance

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
        else:
            print(f'[{self.owner} 계좌] 출금 실패: 잔액이 부족합니다.')
        return self.balance
````

````python
# 철수와 영희 계좌를 생성하고 입출금을 수행하세요.
account_1 = BankAccount('김철수', 5000)
account_2 = BankAccount('이영희', 10000)

account_1.deposit(2000)
account_2.withdraw(3000)

print(f'{account_1.owner}님 잔액: {account_1.balance}')
print(f'{account_2.owner}님 잔액: {account_2.balance}')
````

## 심화 1. 조건에 따라 상태가 변경되는 객체

### 문제 4-1. `Hero` 체력 관리

### 요구사항

1. `Hero` 클래스를 정의하세요.
2. `__init__(self, name)`에서 `name`을 저장하고 `hp = 100`으로 초기화하세요.
3. `take_damage(self, damage)` 메서드를 작성하세요.
4. 피해량만큼 `hp`를 감소시키세요.
5. 체력은 0보다 작아지지 않도록 처리하세요.

```python
self.hp = max(0, self.hp - damage)
```

6. 피해 처리 후 다음 형식으로 출력하세요.

```text
용사의 남은 체력: 70
```

7. 체력이 0이 되면 다음 문구를 추가로 출력하세요.

```text
용사 캐릭터가 쓰러졌습니다.
```

8. 피해량 `30`, `50`, `40`을 순서대로 적용하세요.

````python
# Hero 클래스를 작성하세요.
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100 #호출할 때마다 100으로 고정되어있어야 하기 때문에 매개 변수로 지정하지 않음

    def take_damage(self, damage):
        self.hp = max(0, self.hp - damage)
        print(f'용사의 남은 체력: {self.hp}')
        if self.hp == 0:
            print('용사 캐릭터가 쓰러졌습니다.')
````

````python
# Hero 객체를 생성하고 피해량을 순서대로 적용하세요.
hero = Hero('용사')

hero.take_damage(30)
hero.take_damage(50)
hero.take_damage(40)
````

### 예상 상태 변화

```text

```

> 강의자료에는 `__repr__`도 소개되지만, 이번 실습에서는 난이도를 고려하여 `__str__`만 직접 구현합니다.

````python
# __str__을 포함한 BankAccount 클래스를 작성하거나 기존 클래스를 수정하세요.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def __str__(self):
        return f'{self.owner}님의 계좌 (현재 잔액: {self.balance})'
````
