## 필수 1. `BankAccount`를 상속한 `MinusAccount`

### 문제 1-1. 마이너스 통장 클래스 구현

8장 1강의 일반 계좌를 확장하여 잔액이 부족해도 설정된 마이너스 한도까지 출금할 수 있도록 구현합니다.

### 요구사항

1. 다음 부모 클래스를 작성하세요.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"[{self.owner} 일반 계좌] {amount}원 출금 완료 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 일반 계좌] 출금 실패: 잔액이 부족합니다.")
        return self.balance
```

2. `BankAccount`를 상속받는 `MinusAccount` 클래스를 정의하세요.
3. `MinusAccount.__init__(self, owner, balance, limit)`을 작성하세요.
4. 부모의 `owner`, `balance` 초기화는 다음 방식으로 재사용하세요.

```python
super().__init__(owner, balance)
```

5. 자식 클래스에 `limit` 속성을 추가하세요.
6. `withdraw(self, amount)`를 오버라이딩하세요.
7. 다음 조건일 때 출금을 허용하세요.

```python
(self.balance + self.limit) >= amount
```

8. 출금 성공 시 `balance`에서 출금액을 차감하세요.
9. 한도를 초과하면 잔액을 변경하지 않고 실패 메시지를 출력하세요.
10. 다음 계좌를 생성하세요.

```python
minus_account = MinusAccount("홍길동", 5000, 10000)
```

11. `12000`원을 출금하여 마이너스 잔액이 되는지 확인하세요.
12. 이어서 한도를 초과하는 금액을 출금하여 실패하는지 확인하세요.

````python
# BankAccount와 MinusAccount 클래스를 작성하세요.
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            print(f"[{self.owner} 일반 계좌] {amount}원 출금 완료 / 현재 잔액: {self.balance}원")
        else:
            print(f"[{self.owner} 일반 계좌] 출금 실패: 잔액이 부족합니다.")
        return self.balance

class MinusAccount(BankAccount):
    def __init__(self, owner, balance, limit):
        super().__init__(owner, balance)
        self.limit = limit
    def withdraw(self, amount):
        if (self.balance + self.limit) >= amount:
            self.balance -= amount
        else:
            print(f"[{self.owner} 마이너스 계좌] 출금 실패: 한도 초과")

````

````python
# MinusAccount 인스턴스를 생성하고 출금을 테스트하세요.
minus_account = MinusAccount("홍길동", 5000, 10000)
minus_account.withdraw(12000)
minus_account.withdraw(6000)
````

```text

```

테스트:

```text

```

두 번째 출금 시 현재 출금 가능 여유는 3000원이므로 실패해야 합니다.

## 필수 3. `ProtectedAccount`와 Getter·Setter

### 문제 3-1. 잔액 직접 변경 제한과 유효성 검증

### 요구사항

1. `ProtectedAccount` 클래스를 정의하세요.
2. `__init__(self, owner, balance)`에서 다음 속성을 저장하세요.

```python
self.owner = owner
self.__balance = balance
```

3. 내부 잔액을 반환하는 `get_balance(self)`를 정의하세요.
4. `set_balance(self, amount)`를 정의하세요.
5. `amount >= 0`일 때만 `__balance`를 변경하세요.
6. 정상 변경 시 다음 형식으로 출력하세요.

```text
[시스템] 잔액이 5000원으로 정상 변경되었습니다.
```

7. 음수 값이면 기존 잔액을 유지하고 다음 문구를 출력하세요.

```text
[시스템 경고] 잘못된 수치 입력: 잔액은 음수가 될 수 없습니다.
```

8. 다음 객체를 생성하세요.

```python
acc = ProtectedAccount("이몽룡", 3000)
```

9. Getter로 초기 잔액을 확인하세요.
10. `set_balance(-1000)`을 실행해 변경이 차단되는지 확인하세요.
11. `set_balance(5000)`을 실행해 정상 변경되는지 확인하세요.
12. Getter로 최종 잔액을 출력하세요.

````python
# ProtectedAccount 클래스를 작성하세요.
class ProtectedAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
            print(f'[시스템] 잔액이 {self.__balance}원으로 정상 변경되었습니다.')
        else:
            print('[시스템 경고] 잘못된 수치 입력: 잔액은 음수가 될 수 없습니다.')
````

````python
# Getter와 Setter의 검증 흐름을 테스트하세요.
acc = ProtectedAccount('이몽룡', 3000)

print(f'초기잔액: {acc.get_balance()}')

acc.set_balance(-1000)
print(f'실패 후 잔액: {acc.get_balance()}')

acc.set_balance(5000)
print(f'최종 잔액: {acc.get_balance()}')
````
