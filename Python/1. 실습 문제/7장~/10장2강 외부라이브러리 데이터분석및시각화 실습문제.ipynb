## 심화 1. Pandas와 Matplotlib을 결합한 매출 차트

### 문제 4-1. 월별 매출 추이 완성형 차트

필수 1에서 만든 `df_sales`를 사용하세요.

### 요구사항
1. `df_sales["월"]`을 X축 데이터로 사용하세요.
2. `df_sales["매출"]`을 Y축 데이터로 사용하세요.
3. `plt.plot()`에 다음 옵션을 적용하세요.

```python
marker="o"
label="Monthly Sales"
```

4. 차트 제목을 `"Sparta Lecture Sales Report"`로 지정하세요.
5. X축 이름을 `"Month"`로 지정하세요.
6. Y축 이름을 `"Sales (KRW)"`로 지정하세요.
7. `plt.grid(True)`로 격자를 추가하세요.
8. `plt.legend()`로 범례를 표시하세요.
9. `plt.show()`로 그래프를 출력하세요.

> 강의 예시에는 `color="green"`이 있지만 색상 지정은 이번 실습의 필수 평가 요소가 아닙니다.

````python
# df_sales의 월/매출 열을 사용해 완성형 선 그래프를 작성하세요.
x = df_sales["월"]
y = df_sales["매출"]

plt.plot(x, y, marker = "o", label = "Monthly Sales")
plt.title("Sparta Lecture Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales (KRW)")
plt.grid(True)
plt.legend()

plt.show()
````
