## 필수 2. 표준 라이브러리 활용

### 문제 2-1. 이벤트 번호와 실행 시각 기록

### 요구사항
1. `random` 모듈을 불러오세요.
2. `random.randint(1, 45)`로 이벤트 번호를 생성하세요.
3. `datetime` 모듈을 불러오세요.
4. `datetime.datetime.now()`로 현재 시각 객체를 생성하세요.
5. `strftime("%Y-%m-%d %H:%M:%S")`로 문자열을 만드세요.
6. 다음 형식으로 출력하세요.

```text
추첨 번호: 23
실행 시각: 2026-08-07 10:20:00
```

> 번호와 시각은 실행할 때마다 달라질 수 있습니다.

````python
# random과 datetime 표준 모듈을 활용하세요.
import random
import datetime

lotte = random.randint(1, 45)
current_time = datetime.datetime.now()

format_time = current_time.strftime("%Y-%m-%d %H:%M:%S")

print(f"추첨 번호: {lotte}")
print(f"실행 시각: {format_time}")
````

### 확인 사항
- 표준 라이브러리는 파이썬 설치 시 함께 제공되므로 별도 설치가 필요하지 않습니다.
- `randint(1, 45)`는 1과 45를 모두 포함합니다.
- `%Y-%m-%d %H:%M:%S`는 연-월-일 시:분:초 형식입니다.

## 필수 3. `import`, `from`, `as` 방식 비교

### 문제 3-1. 같은 기능을 서로 다른 방식으로 호출하기

### 요구사항
1. `import math`를 사용하여 `math.sqrt(16)`을 출력하세요.
2. `from math import sqrt`를 사용하여 `sqrt(25)`를 출력하세요.
3. `import math as m`을 사용하여 `m.sqrt(36)`을 출력하세요.
4. 결과를 각각 다음 형식으로 출력하세요.

```text
import 방식: 4.0
from 방식: 5.0
as 방식: 6.0
```

5. 세 방식의 장단점을 한 문장씩 작성하세요.

````python
# import, from, as 방식을 각각 사용하세요.
import math 
print(f"import 방식: {math.sqrt(16)}")

from math import sqrt
print(f"from 방식: {sqrt(25)}")

import math as m
print(f"as 방식: {m.sqrt(36)}")
````

### 확인 사항
- `import 모듈명`: 출처가 명확합니다.
- `from 모듈 import 이름`: 호출이 짧지만 이름 충돌 가능성을 고려해야 합니다.
- `import 모듈 as 별칭`: 긴 모듈 이름을 줄이거나 관례적인 별칭을 사용할 수 있습니다.

### 패키지 모듈 실행 방식
