## 필수 1. 딕셔너리 수정과 안전한 키 조회

### 문제 1-1. 회원 정보 갱신 및 연락처 기본값 처리

### 요구사항

1. 이름 `"Sparta"`와 나이 `20`을 갖는 `user_info` 딕셔너리를 생성하세요.
2. `"age"` 키의 값을 `21`로 수정하세요.
3. `"role"` 키를 새로 추가하고 `"admin"`을 저장하세요.
4. `get("phone")`을 사용하여 존재하지 않는 연락처를 조회하고 결과를 `phone_without_default`에 저장하세요.
5. 연락처가 등록되어 있지 않은 상황에서 `get("phone")`을 사용하여 조회 결과를 `phone_without_default`에 저장하세요.
6. `get("phone", "010-0000-0000")`을 사용하여 연락처가 없을 때 사용할 기본값을 `phone_with_default`에 저장하세요.

### 예상 출력

```text
{'name': 'Sparta', 'age': 21, 'role': 'admin'}
None
010-0000-0000
```

````python
# 문제 1-1 코드를 작성하세요.
user_info = {"name":"Sparta", "age":20}
user_info["age"] = 21
user_info["role"] = "admin"
print(user_info)

print(user_info.get("phone"))
phone_without_default = user_info.get("phone")

phone_with_default = user_info.get("phone", "010-0000-0000")
print(phone_with_default)
````

### 확인 사항

- 동일한 `"age"` 키가 두 개 생기는 것이 아니라 기존 값이 갱신되는지 확인하세요.
- `get()`에 기본값을 지정하지 않으면 `None`이 반환되는지 확인하세요.
- 대괄호로 없는 키를 직접 조회하지 않았는지 확인하세요.

## 필수 2. 집합을 활용한 이메일 중복 제거

### 문제 2-1. 수집 이메일 정제

### 요구사항

1. 아래 이메일 목록을 `raw_emails`에 저장하세요.

```python
["a@sparta.com", "b@sparta.com", "a@sparta.com"]
```

2. `set()`으로 변환하여 `unique_emails`를 생성하세요.
3. `list()`로 다시 변환하여 `refined_emails`를 생성하세요.
4. 원본 이메일 목록, 집합, 변환된 리스트를 각각 출력하세요.
5. 중복 제거 전후의 요소 개수를 `len()`으로 출력하세요.

### 예상 핵심 결과

```text
중복 제거 전 개수: 3
중복 제거 후 개수: 2
```

> `refined_emails`는 리스트이므로 인덱싱할 수 있지만, 원본의 입력 순서가 유지된다고 보장할 수는 없습니다.

````python
# 문제 2-1 코드를 작성하세요.
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]
unique_emails = set(raw_emails)
refined_emails = list(unique_emails)

print(raw_emails)
print(unique_emails)
print(refined_emails)

print(len(raw_emails))
print(len(refined_emails))
````

### 확인 사항

- 중복된 `"a@sparta.com"`이 한 번만 남았는지 확인하세요.
- 집합에서 리스트로 변환해도 원래 순서가 복구되는 것은 아닙니다.

## 필수 3. 딕셔너리 뷰 메서드 활용

### 문제 3-1. 메뉴 데이터 구성 확인

### 요구사항

1. 다음 딕셔너리를 `menu`에 저장하세요.

```python
{"coffee": 4000, "tea": 4500}
```

2. `keys()`로 메뉴명 목록을 출력하세요.
3. `values()`로 가격 목록을 출력하세요.
4. `items()`로 키-값 쌍을 출력하세요.
5. `get("juice", 5000)`을 사용하여 없는 메뉴의 기본 가격을 출력하세요.

### 예상 출력 형태

```text
dict_keys(['coffee', 'tea'])
dict_values([4000, 4500])
dict_items([('coffee', 4000), ('tea', 4500)])
5000
```

````python
# 문제 3-1 코드를 작성하세요.
menu = {"coffee": 4000, "tea": 4500}

print(menu.keys())
print(menu.values())
print(menu.items())

price = menu.get("juice", 5000)
print(price)
````

## 심화 1. 집합 연산과 포함 여부 확인

### 문제 4-1. 캠페인 참여 고객 비교

### 요구사항

1. `campaign_a`에 `{101, 102, 103}`을 저장하세요.
2. `campaign_b`에 `{103, 104, 105}`를 저장하세요.
3. `&`로 두 캠페인에 모두 참여한 고객을 출력하세요.
4. `|`로 한 번 이상 참여한 전체 고객을 출력하세요.
5. `campaign_a - campaign_b`로 A에만 참여한 고객을 출력하세요.
6. `campaign_b - campaign_a`로 B에만 참여한 고객을 출력하세요.

### 예상 핵심 결과

```text
{103}
{101, 102, 103, 104, 105}
{101, 102}
{104, 105}
True
True
```

> 집합 요소의 출력 순서는 달라질 수 있습니다.

````python
# 문제 4-1 코드를 작성하세요.
campaign_a = {101, 102, 103}
campaign_b = {103, 104, 105}

all_users = campaign_a & campaign_b
print(all_users)

users = campaign_a | campaign_b
print(users)

only_a = campaign_a - campaign_b
print(only_a)

only_b = campaign_b - campaign_a
print(only_b)
````

### 확인 사항

- `&`는 공통 고객, `|`는 전체 고객, `-`는 한쪽에만 속한 고객을 찾습니다.
- 집합의 포함 여부 검사는 일반적으로 리스트보다 빠르게 동작하는 편입니다.
