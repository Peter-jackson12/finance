함수 밖의 전역 포인트와 함수 내부의 임시 포인트가 서로 독립적으로 유지되는지 확인합니다.

### 요구사항

2. `reward_service` 함수를 정의하세요.
3. 함수 내부에 같은 이름의 지역 변수 `user_point = 100`을 선언하세요.
4. 함수 내부 값을 다음 형식으로 출력하세요.

```text
함수 내부 포인트: 100
```

## 필수 2. `global` 방식과 `return` 방식 비교

### 문제 2-1. 처리 건수 누적

전역 변수를 직접 수정하는 방식과 값을 입력받아 반환하는 방식을 각각 구현합니다.

### 1단계: `global` 사용

1. 전역 변수 `processed_count = 0`을 선언하세요.
2. `increase_count_global` 함수를 정의하세요.
3. 함수 내부에서 `global processed_count`를 선언하세요.
4. 처리 건수를 1 증가시키세요.
5. 함수를 두 번 호출하고 최종 값을 출력하세요.

### 2단계: `return` 사용

1. `increase_count` 함수를 정의하세요.
2. 현재 건수를 매개변수 `count`로 받으세요.
3. `count + 1`을 반환하세요.
4. `safe_count = 0`을 선언하세요.
5. 반환값을 다시 `safe_count`에 저장하는 방식으로 두 번 호출하세요.
6. 최종 값을 출력하세요.

### 예상 결과

```text
global 방식: 2
return 방식: 2
```

````python
# 1단계: global 방식으로 작성하세요.
processed_count = 0

def increase_count_global():
    global processed_count
    processed_count += 1

increase_count_global()
increase_count_global()
print(processed_count)
````

````python
# 2단계: return 방식으로 작성하세요.
def increase_count(count):
    return count + 1

safe_count = 0
safe_count = increase_count(safe_count)
safe_count = increase_count(safe_count)

print("return:", safe_count)
````

### 확인 사항

- 두 방식 모두 결과는 같지만 `return` 방식은 입력과 출력이 명확합니다.
- 전역 변수는 여러 함수가 함께 수정하면 변경 원인을 추적하기 어려울 수 있습니다.

## 필수 3. 일반 함수와 람다 함수 비교

### 문제 3-1. 상품 가격에 5,000원 추가

같은 계산을 일반 함수와 람다 함수로 각각 작성합니다.

### 요구사항

1. `add_fee` 일반 함수를 정의하여 입력값에 `5000`을 더해 반환하세요.
2. 같은 계산을 수행하는 람다 함수를 `lambda_add_fee`에 저장하세요.
3. 두 함수에 각각 `20000`을 전달하여 결과를 출력하세요.
4. 두 결과가 같은지 비교하여 출력하세요.

### 예상 결과

```text
25000
25000
True
```

````python
# 일반 함수와 람다 함수를 작성하고 결과를 비교하세요.
def add_fee(price):
    return price + 5000

lambda_add_fee = lambda x: x + 5000

print(add_fee(20000))
print(lambda_add_fee(20000))
print(add_fee(20000) == lambda_add_fee(20000))
````

### 확인 사항

- 람다는 `lambda 매개변수: 표현식` 형태로 작성합니다.
- 람다는 `return`을 직접 작성하지 않아도 표현식 결과를 반환합니다.
- 복잡하거나 여러 단계인 로직은 일반 `def` 함수가 더 적합합니다.

## 심화 1. LEGB 탐색 순서 확인

### 문제 4-1. 설정값 탐색

중첩 함수에서 같은 이름의 변수가 있을 때 가장 가까운 스코프의 값이 선택되는지 확인합니다.

### 요구사항

1. 전역 변수 `setting = "Global 설정"`을 선언하세요.
2. `outer_function`을 정의하세요.
3. 외부 함수 안에 `setting = "Enclosing 설정"`을 선언하세요.
4. 내부 함수 `inner_function`을 정의하세요.
5. 내부 함수에 `setting = "Local 설정"`을 선언하고 출력하세요.
6. 내부 함수 호출 후 외부 함수의 `setting`도 출력하세요.
7. 외부 함수 실행 후 전역 `setting`도 출력하세요.

### 예상 결과

```text
Local 설정
Enclosing 설정
Global 설정
```

````python
# Local, Enclosing, Global 값을 확인하는 코드를 작성하세요.
setting = "Global 설정"

def outer_function():
    setting = "Enclosing 설정"

    def inner_function():
        setting = "Local 설정"
        print(setting)

    inner_function()
    print(setting)

outer_function()
print(setting)

````

### 확인 사항

- 파이썬은 Local → Enclosing → Global → Built-in 순서로 이름을 찾습니다.
- 현재 문제에서는 L, E, G 세 범위를 직접 확인합니다.
- `print`, `len` 같은 내장 이름은 Built-in 범위에 있습니다.

## 심화 2. 람다와 데이터 가공 함수 연동

### 문제 5-1. 주문 금액 변환 및 고액 주문 필터링

주문 금액에 10%를 추가한 뒤, 변환 금액이 50,000원을 초과하는 주문만 선별합니다.

### 요구사항

1. `order_amounts = [20000, 35000, 50000, 70000]`을 선언하세요.
2. `map()`과 `lambda`를 사용해 각 금액에 10%를 추가하세요.
3. 계산 결과는 정수로 변환하여 `adjusted_amounts`에 저장하세요.
4. `filter()`와 `lambda`를 사용해 50,000원을 초과하는 값만 선별하세요.
5. 결과를 `high_value_orders`에 저장하세요.
6. 두 리스트를 각각 출력하세요.

### 예상 결과

```text
[22000, 38500, 55000, 77000]
[55000, 77000]
```

````python
# map()과 lambda로 금액을 변환하세요.
order_amounts = [20000, 35000, 50000, 70000]

adjusted_amounts = list(map(lambda amount: int(amount * 1.1), order_amounts))

print(adjusted_amounts)
````

````python
# filter()와 lambda로 고액 주문을 선별하세요.
high_value_orders = list(filter(lambda high: high > 50000, adjusted_amounts))

print(high_value_orders)
````

### 확인 사항

- `map()`은 모든 요소에 동일한 변환 규칙을 적용합니다.
- `filter()`는 조건식이 참인 원소만 남깁니다.
- 람다는 짧은 일회성 규칙을 함수 인수로 전달할 때 유용합니다.
