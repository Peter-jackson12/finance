# [5장 3강] 클로저와 데코레이터 — 최종 실습문제

## 실습 목표

- 함수의 일급 객체 특성을 활용하여 함수를 변수에 저장하고 다른 함수에 전달하거나 반환할 수 있다.
- 클로저가 외부 함수의 변수를 기억하는 구조를 설명하고 `nonlocal`로 상태를 변경할 수 있다.
- 데코레이터의 수동 연결 구조와 `@` 문법의 관계를 이해할 수 있다.
- 데코레이터로 원본 함수의 실행 전후에 로깅과 실행 시간 측정 기능을 추가할 수 있다.
- `functools.wraps`와 반환값 처리를 통해 원본 함수의 기능과 정보를 유지할 수 있다.

## 실습 환경

- Python 3.14
- Jupyter Notebook
- Python 표준 라이브러리: `functools`, `time`

## 실무 연결

클로저는 전역 변수 없이 개별 상태를 유지하는 간단한 저장소나 설정 함수에 활용할 수 있습니다.  
데코레이터는 여러 업무 함수에 공통으로 필요한 로그 기록, 실행 시간 측정, 권한 검사 등의 기능을 원본 코드와 분리할 때 활용합니다.

> 5장 1강의 기본 함수 문법과 5장 2강의 일반적인 스코프·람다 설명은 반복하지 않습니다.
