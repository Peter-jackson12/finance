
## 필수 2. 기본값 매개변수와 다중 반환

### 문제 2-1. 주문 결제 금액 계산

### 요구사항

1. `calculate_total` 함수를 정의하세요.
2. 매개변수는 `price`, `delivery_fee=3000` 순서로 작성하세요.
3. 상품 가격과 배송비를 더한 값을 `total`에 저장하세요.
4. `total`과 `"결제가 완료되었습니다"`를 동시에 반환하세요.
5. `calculate_total(25000)`을 호출하여 기본 배송비가 적용되도록 하세요.
6. 반환값을 `total_price`, `payment_message`에 언패킹하고 출력하세요.
7. `calculate_total(25000, 5000)`을 호출하여 배송비를 직접 변경한 결과도 출력하세요.
8. 반환값을 하나의 변수 `all_result`로도 받아서 값과 자료형을 출력하세요.

### 예상 핵심 결과

```text
28000
결제가 완료되었습니다
30000
결제가 완료되었습니다
(28000, '결제가 완료되었습니다')
<class 'tuple'>
```

````python
# calculate_total 함수를 정의하세요.
def calculate_total(price, delivery_fee=3000):
    total = price + delivery_fee
    return total, "결제가 완료되었습니다."
calculate_total(25000)
````

````python
# 기본 배송비를 사용하고 두 변수로 언패킹하세요.
total_price, payment_message = calculate_total(25000)
print(total_price)
print(payment_message)
````

````python
# 배송비 5000원을 직접 전달하세요.
calculate_total(25000, 5000)
````

````python
# 반환값을 하나의 변수로 받고 자료형을 확인하세요.
all_result = calculate_total(25000)
print(all_result)
type(all_result)
````

### 확인 사항

- 기본값이 없는 매개변수는 기본값 매개변수보다 앞에 있어야 합니다.
- 여러 값을 쉼표로 반환하면 실제로는 하나의 튜플로 묶여 반환됩니다.
- `print()`는 화면 출력, `return`은 함수 밖으로 결과 전달입니다.

## 필수 3. 가변 인수로 데이터 집계

### 문제 3-1. 과목 점수 총점과 평균 계산

### 요구사항

1. `get_total_and_average` 함수를 정의하세요.
2. 매개변수는 `*scores`로 작성하세요.
3. 전달된 점수가 없으면 `0, 0`을 즉시 반환하세요.
4. 점수가 있으면 `sum()`으로 총점을 계산하세요.
5. `len()`으로 점수 개수를 구해 평균을 계산하세요.
6. 총점과 평균을 동시에 반환하세요.
7. 다음 세 경우를 호출하고 출력하세요.
   - `90, 85, 95`
   - `70, 80, 90, 100, 65`
   - 인수를 전달하지 않은 경우

### 예상 핵심 결과

```text
3과목 결과: 270 90.0
5과목 결과: 405 81.0
점수 없음 결과: 0 0
```

````python
# get_total_and_average 함수를 정의하세요.
def get_total_and_average(*scores):
    if len(scores) == 0:
        return 0, 0

    total = sum(scores)
    average = total / len(scores)
    return total, average


````

````python
# 3과목 점수로 호출하세요.
total, average = get_total_and_average(90, 85, 95)
print("3과목 결과:", total, average)
````

````python
# 5과목 점수로 호출하세요.
total, average = get_total_and_average(70, 80, 90, 100, 65)
print("5과목 결과: ", total, average)
````

````python
# 인수 없이 호출하세요.
total, average = get_total_and_average()
print("점수 없음 결과: ", total, average)
````

### 확인 사항

- `scores`는 함수 내부에서 튜플로 전달됩니다.
- 점수가 없는 경우를 먼저 처리하지 않으면 0으로 나누는 오류가 발생할 수 있습니다.
- `return 0, 0`이 실행되면 함수는 즉시 종료됩니다.

## 심화 1. `**kwargs`로 회원 프로필 구성

### 문제 4-1. 항목 수가 유동적인 회원 정보 수집

### 요구사항

1. `create_user_profile` 함수를 정의하세요.
2. 매개변수는 `**user_data`로 작성하세요.
3. 함수 내부에서 `user_data`와 자료형을 출력하세요.
4. 전달된 딕셔너리를 그대로 반환하세요.
5. 아래 정보를 키워드 인수로 전달하세요.
   - `name="스파르타"`
   - `level=5`
   - `active=True`
6. 반환값을 `profile`에 저장하고 출력하세요.

### 예상 핵심 결과

```text
{'name': '스파르타', 'level': 5, 'active': True}
<class 'dict'>
```

````python
# create_user_profile 함수를 정의하세요.
def create_user_profile(**user_data):
    print(type(user_data))
    for key, value in user_data.items():
        print(f" - {key}: {value}")
        
````

````python
# 회원 정보를 전달하고 반환값을 확인하세요.
create_user_profile(name = "스파르타", level = 5, active = True)
````

### 확인 사항

- `*args`는 여러 위치 인수를 튜플로 받습니다.
- `**kwargs`는 여러 키워드 인수를 딕셔너리로 받습니다.

## 심화 2. 조건문을 포함한 안전한 함수

### 문제 5-1. 안전한 나누기 기능 분리

조건문과 함수를 결합하여 0으로 나누는 상황을 안전하게 처리합니다.

### 요구사항

1. `divide` 함수를 정의하세요.
2. 매개변수는 `a`, `b`로 작성하세요.
3. `b`가 `0`이면 `"0으로 나눌 수 없습니다."`를 즉시 반환하세요.
4. 그렇지 않으면 `a / b` 결과를 반환하세요.
5. `divide(20, 4)`와 `divide(20, 0)`을 각각 호출하여 출력하세요.

### 예상 결과

```text
5.0
0으로 나눌 수 없습니다.
```

````python
# divide 함수를 정의하세요.
def divide(a, b):
    if b == 0:
        return "0으로 나눌 수 없습니다."
    return a / b
````

````python
# 정상 나누기와 0으로 나누는 경우를 테스트하세요.
print(divide(20, 4))

print(divide(20, 0))
````

### 확인 사항

- 오류가 발생하기 전에 조건문으로 위험한 입력을 검사했는지 확인하세요.
- 첫 번째 `return`이 실행되면 아래 나눗셈 코드는 실행되지 않습니다.
- 기능을 함수로 분리하면 다른 코드에서도 같은 검증 로직을 재사용할 수 있습니다.
