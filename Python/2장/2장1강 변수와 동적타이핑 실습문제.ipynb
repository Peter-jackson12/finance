## 필수 3. 주피터 노트북의 셀 간 변수 공유 확인하기

### 문제 3-1. 고객 점수 누적하기

1. 첫 번째 셀에서 `customer_score`에 `70`을 저장하세요.
2. 두 번째 셀에서 기존 점수에 `10`을 더해 다시 저장하세요.
3. 세 번째 셀에서 기존 점수에 `5`를 더해 다시 저장하세요.

````python
# 1단계: 초기 고객 점수를 저장하세요.

customer_score = 70
````

````python
# 2단계: 기존 점수에 10을 더해 재할당하세요.

customer_score+= 10
````

````python
# 3단계: 기존 점수에 5를 더해 재할당하세요.
customer_score+= 5

customer_score
````
