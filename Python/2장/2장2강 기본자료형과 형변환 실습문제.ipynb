## 필수 1. 네 가지 기본 자료형 식별하기

### 문제 1-1. 회원 및 주문 정보의 자료형 확인

온라인 쇼핑몰의 회원 정보와 주문 상태를 코드에 저장했습니다.  
각 값이 파이썬에서 어떤 자료형으로 처리되는지 확인하세요.

### 요구사항

1. `member_name`에 문자열 `"김민수"`를 저장하세요.
2. `member_age`에 정수 `20`을 저장하세요.
3. `order_amount`에 실수 `12500.5`를 저장하세요.
4. `payment_completed`에 불리언 값으로 불리는 `True`를 저장하세요.
5. 네 변수의 값과 자료형을 각각 출력하세요.
6. 다음 값들을 `bool()`로 변환하여 결과를 출력하세요.
   - `0`
   - `""`
   - `1`
   - `"Python"`

### 예상 핵심 결과

```text
<class 'str'>
<class 'int'>
<class 'float'>
<class 'bool'>
False
False
True
True
```

````python
# 문제 1-1 코드를 작성하세요.

member_name = "김민수"
member_age = 20
order_amount = 12500.5
payment_completed = True

print(type(member_name))
print(type(member_age))
print(type(order_amount))
print(type(payment_completed))

print(member_name, type(member_name))

print(bool(0))
print(bool(""))
print(bool(1))
print(bool("Python"))

print(bool("False"))
````

### 확인 사항

- `True`의 첫 글자를 대문자로 작성했는지 확인하세요.
- 숫자 `0`과 빈 문자열은 `False`, 값이 있는 숫자와 문자열은 `True`로 변환되는지 확인하세요.

## 필수 2. 명시적 형변환으로 주문 데이터 가공하기

### 문제 2-1. 문자열 주문 수량을 숫자로 변환하기

웹 화면에서 입력받은 주문 수량과 상품 가격은 문자열 형태로 전달되었다고 가정합니다.  
계산을 수행하려면 먼저 숫자 자료형으로 변환해야 합니다.

### 요구사항

1. `quantity_text`에 문자열 `"3"`을 저장하세요.
2. `price_text`에 문자열 `"4500.5"`를 저장하세요.
3. `quantity_text`를 `int()`로 변환하여 `quantity`에 저장하세요.
4. `price_text`를 `float()`로 변환하여 `price`에 저장하세요.
5. `quantity * price`를 계산하여 `total_amount`에 저장하세요.
6. 변환된 두 값의 자료형과 총금액을 출력하세요.
7. `quantity`를 `str()`로 변환하여 다음 안내 문장을 출력하세요.

```text
주문 수량은 3개입니다.
```

### 예상 핵심 결과

```text
<class 'int'>
<class 'float'>
13501.5
주문 수량은 3개입니다.
```

````python
# 문제 2-1 코드를 작성하세요.

quantity_text = '3'
price_text = "4500.5"
quantity = int(quantity_text)
price = float(price_text)
total_amount = quantity * price
print(type(quantity))
print(type(price))
print(total_amount)

message = "주문 수량은 " + str(quantity) + "개입니다."
print(message)
````

### 확인 사항

- 문자열 상태에서 곱셈 계산을 시도하지 않았는지 확인하세요.
- 정수형 변환에는 `int()`, 실수형 변환에는 `float()`를 사용했는지 확인하세요.
- 숫자와 문자열을 결합할 때 `str()`을 사용했는지 확인하세요.

## 필수 3. 묵시적 형변환과 소수점 처리 확인하기

### 문제 3-1. 기본 금액과 추가 수수료 계산

기본 금액은 정수형이고 추가 수수료는 실수형으로 저장되어 있습니다.  
두 값을 더할 때 결과 자료형과 정수형 변환 결과를 확인하세요.

### 요구사항

1. `base_amount`에 정수 `10`을 저장하세요.
2. `additional_fee`에 실수 `3.14`를 저장하세요.
3. 두 값을 더해 `result`에 저장하세요.
4. `result`의 값과 자료형을 출력하세요.
5. `int(result)`를 `converted_result`에 저장하세요.
6. 변환된 값과 자료형을 출력하세요.

### 예상 출력

```text
13.14
<class 'float'>
13
<class 'int'>
```

````python
# 문제 3-1 코드를 작성하세요.

base_amount = 10
additional_fee = 3.14

result = base_amount + additional_fee
print(result)
print(type(result))

converted_result = int(result)
print(converted_result)
print(type(converted_result))
````

### 확인 사항

- 정수와 실수를 연산한 결과가 `float`인지 확인하세요.
- `int()`는 반올림하지 않고 소수점 이하를 제거한다는 점을 확인하세요.
