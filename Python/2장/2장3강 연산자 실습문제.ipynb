## 필수 1. 산술 연산자와 복합 대입 연산자로 재고 갱신하기

### 문제 1-1. 상품 재고 변동 계산

상품의 초기 재고가 입고와 판매에 따라 변경되는 상황입니다.

### 요구사항

1. `stock_count`에 초기 재고 `10`을 저장하세요.
2. `*=`를 사용하여 재고를 5배로 늘리고 출력하세요.
3. `-=`를 사용하여 판매 수량 `8`을 차감하고 출력하세요.
4. `+=`를 사용하여 추가 입고 수량 `3`을 더하고 출력하세요.

### 예상 출력

```text
50
42
45
```

````python
# 문제 1-1 코드를 작성하세요.

stock_count = 10
stock_count *= 5
stock_count -= 8
stock_count += 3
print(stock_count)
````

## 필수 2. 주문 금액과 포장 단위 계산하기

### 문제 2-1. 산술 연산자 종합 활용

상품 단가와 주문 수량을 이용하여 금액과 포장 정보를 계산합니다.

### 요구사항

1. `unit_price`에 `3500`, `order_quantity`에 `23`을 저장하세요.
2. `*`를 사용하여 총 주문 금액을 계산하여 `total_price`변수에 저장해주세요
3. 상품을 5개씩 포장할 때 완성되는 상자 수를 `//`로 계산하세요.
4. 포장 후 남는 상품 수를 `%`로 계산하세요.
5. 총 주문 금액을 주문 수량으로 나눈 값을 `/`로 계산하고 자료형도 출력하세요.
6. `2 ** 3`을 계산하여 2개 선택지가 3단계에서 만들 수 있는 조합 수를 출력하세요.

### 예상 핵심 결과

```text
80500
4
3
3500.0
<class 'float'>
8
```

````python
# 문제 2-1 코드를 작성하세요.

unit_price = 3500
order_quantity = 23

total_price = unit_price * order_quantity

print(total_price)
print(order_quantity // 5)
print(order_quantity % 5)
print(total_price / order_quantity)
print(type(total_price / order_quantity))
print(2**3)
````

### 확인 사항

- `/`는 정수끼리 계산해도 결과가 `float`인지 확인하세요.
- `//`는 몫, `%`는 나머지, `**`는 거듭제곱을 반환합니다.

## 필수 3. 비교 연산자로 주문 데이터 판별하기

### 문제 3-1. 주문 금액과 상태 비교

### 요구사항

1. `order_amount`에 `120000`을 저장하세요.
2. `order_status`에 문자열 `"결제완료"`를 저장하세요.
3. 주문 금액이 `100000` 이상인지 `>=`로 비교하세요.
4. 주문 상태가 `"결제완료"`와 같은지 `==`로 비교하세요.
5. 주문 상태가 `"취소"`와 다른지 `!=`로 비교하세요.
6. `"apple"`과 `"banana"`를 `<`로 비교하세요.
7. 네 결과를 각각 출력하세요.

### 예상 출력

```text
True
True
True
True
```

````python
# 문제 3-1 코드를 작성하세요.

order_amount = 120000
order_status = "결제완료"

print(order_amount >= 100000)
print(order_status == "결제완료")
print(order_status != "취소")
print("apple" < "banana")
````

## 필수 4. 논리 연산자로 주문 가능 여부 판단하기

### 문제 4-1. 회원 주문 및 무료 배송 조건

### 요구사항

1. `order_amount`에 `120000`을 저장하세요.
2. `is_member`에 `True`, `is_blocked`에 `False`를 저장하세요.
3. 회원이고 차단 상태가 아닌지를 `and`와 `not`으로 결합하여 `can_order`에 저장하세요.
4. 주문 금액이 `100000` 이상이고 주문 가능한지를 결합하여 `free_shipping`에 저장하세요.
5. 기존 무료 배송 조건을 충족하거나, 회원 여부와 관계없이 주문 금액이 200000원 이상이면 무료 배송이 가능하도록 or 조건을 작성하세요.
6. 세 결과를 출력하세요.

### 예상 출력

```text
True
True
True
```

````python
# 문제 4-1 코드를 작성하세요.

order_amount = 120000
is_member = True
is_blocked = False

can_order = (is_member and not is_blocked)
free_shipping = (order_amount >= 100000 and can_order)

print(can_order)
print(free_shipping)
print(free_shipping or order_amount >= 200000)
````

## 필수 5. 연산자 우선순위와 괄호 확인하기

### 문제 5-1. 같은 숫자, 다른 계산 순서

### 요구사항

1. 아래 식을 그대로 계산하여 `result_without_parentheses`에 저장하세요.

```python
10 + 2 * 5 > 25
```

2. 아래 식을 계산하여 `result_with_parentheses`에 저장하세요.

```python
(10 + 2) * 5 > 25
```

3. 두 결과를 출력하세요.
4. 곱셈과 괄호 중 무엇이 먼저 계산되었는지 결과를 통해 확인하세요.

### 예상 출력

```text
False
True
```

````python
# 문제 5-1 코드를 작성하세요.

result_without_parentheses = 10 + 2 * 5 > 25
result_with_parentheses = (10 + 2) * 5 > 25

print(result_without_parentheses)
print(result_with_parentheses)
````
