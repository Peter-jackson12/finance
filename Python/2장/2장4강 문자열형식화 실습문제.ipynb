## 필수 1. 문자열 형식화 3가지 방식 비교하기

### 문제 1-1. 동일한 메뉴 정보를 세 방식으로 출력

문자열 형식화 기술의 발전 과정을 비교합니다.

### 요구사항

1. `menu_name`에 문자열 `"Americano"`를 저장하세요.
2. `menu_price`에 정수 `4500`을 저장하세요.
3. 다음 문장을 `%` 연산자 방식으로 출력하세요.
4. 같은 문장을 `str.format()` 방식으로 출력하세요.
5. 같은 문장을 f-string 방식으로 출력하세요.

### 출력 문장

```text
상품: Americano | 가격: 4500원
```

### 예상 출력

동일한 문장이 세 줄 출력되어야 합니다.

````python
# 문제 1-1 코드를 작성하세요.

menu_name = "Americano"
menu_price = 4500

print("상품: %s | 가격: %d원" % (menu_name, menu_price))
print("상품: {} | 가격: {}원".format(menu_name, menu_price))
print(f"상품: {menu_name} | 가격: {menu_price}원")
````

### 확인 사항

- `%s`, `%d`의 위치와 변수 자료형이 맞는지 확인하세요.
- `.format()`은 문자열 뒤쪽에 변수를 전달해야 한다는 점을 확인하세요.
- f-string은 문자열 안에서 변수의 위치를 바로 확인할 수 있는지 비교하세요.

## 필수 2. 금액과 할인율의 출력 형식 지정하기

### 문제 2-1. 결제 금액 정보 출력

상품 금액과 할인율을 사람이 읽기 쉬운 형식으로 출력합니다.

### 요구사항

1. `price`에 정수 `15000`을 저장하세요.
2. `discount_rate`에 실수 `0.1`을 저장하세요.
3. 가격에 천 단위 쉼표를 적용하여 출력하세요.
4. 할인율은 `discount_rate * 100`을 사용하여 소수점 첫째 자리까지 출력하세요.
5. 다음 형식으로 한 줄에 출력하세요.

### 출력 예시

```text
금액: 15,000원 | 할인율: 10.0%
```

````python
# 문제 2-1 코드를 작성하세요.

price = 15000
discount_rate = 0.1
print(f"금액: {price:,}원 | 할인율: {discount_rate*100:.1f}%")
````

### 확인 사항

- 금액에 `:,`가 적용되었는지 확인하세요.
- 할인율에 `:.1f`가 적용되었는지 확인하세요.

## 필수 3. 간단한 정산 내역 출력하기

### 문제 3-1. 두 상품의 정산 내역 출력

반복문을 사용하지 않고 두 상품의 가격, 수량, 금액을 각각 출력합니다.

### 요구사항

1. 아래 데이터를 변수에 저장하세요.

   * `item1_name = "Americano"`
   * `item1_price = 4500`
   * `item1_quantity = 2`
   * `item2_name = "Cafe Latte"`
   * `item2_price = 5000`
   * `item2_quantity = 1`
2. 각 상품의 금액을 `가격 × 수량`으로 계산하세요.
3. 상품명, 단가, 수량, 금액을 한 줄씩 출력하세요.
4. 단가와 금액에는 천 단위 쉼표를 적용하세요.
5. 다음 형식으로 두 상품의 정산 내역을 출력하세요.

### 출력 예시

```text
상품명: Americano | 단가: 4,500원 | 수량: 2개 | 금액: 9,000원
상품명: Cafe Latte | 단가: 5,000원 | 수량: 1개 | 금액: 5,000원
```

> 영문 메뉴를 사용하여 글자 너비 차이로 인한 정렬 오차를 줄였습니다.

````python
# 문제 3-1 코드를 작성하세요.

item1_name = "Americano"
item1_price = 4500
item1_quantity = 2
item2_name = "Cafe Latte"
item2_price = 5000
item2_quantity = 1

item1_total = item1_price * item1_quantity
item2_total = item2_price * item2_quantity

print(f"상품명: {item1_name} | 단가: {item1_price:,}원 | 수량: {item1_quantity}개 | 금액: {item1_total:,}원")
print(f"상품명: {item2_name} | 단가: {item2_price:,}원 | 수량: {item2_quantity}개 | 금액: {item2_total:,}원")
````
