# [6장 2강] uv를 활용한 의존성 관리 — 최종 실습문제

## 실습 목표

- `uv init`으로 특정 파이썬 버전의 프로젝트를 생성할 수 있다.
- `uv add`, `uv remove`로 외부 패키지를 추가하거나 제거할 수 있다.
- `pyproject.toml`과 `uv.lock`이 갱신되는 흐름을 확인할 수 있다.
- `uv run`으로 프로젝트 가상환경에서 코드를 실행할 수 있다.
- `.venv` 삭제 후 `uv sync`로 동일한 개발 환경을 복원할 수 있다.
- 협업 시 공유해야 할 파일과 공유하지 않아야 할 폴더를 구분할 수 있다.

## 실습 환경

- VS Code
- Windows PowerShell, 명령 프롬프트 또는 Git Bash
- macOS/Linux 터미널
- uv
- 외부 패키지: `pandas`, `matplotlib`

## 실무

실무에서는 개발자마다 패키지를 임의로 설치하지 않고 프로젝트 명세와 잠금 파일을 공유합니다.  
이번 실습에서는 프로젝트 생성부터 의존성 추가·제거, 실행 검증, 가상환경 복원까지의 전체 흐름을 수행합니다.

> 6장 1강에서 다룬 파이썬 모듈과 패키지의 코드 구조는 반복하지 않습니다. 이번 차시는 외부 패키지와 프로젝트 환경 관리에 집중합니다.

## 핵심 개념

- **PyPI**: 파이썬 오픈소스 패키지를 배포하고 내려받는 중앙 저장소
- **직접 의존성**: 프로젝트에서 직접 사용하는 `pandas`, `matplotlib` 등의 패키지
- **하위 의존성**: 직접 의존성이 내부적으로 필요로 하는 다른 패키지
- **`pyproject.toml`**: 프로젝트가 요구하는 파이썬 버전과 직접 의존성의 명세
- **`uv.lock`**: uv가 해결한 구체적인 패키지 버전과 의존성 관계
- **`.venv`**: 현재 컴퓨터에 실제로 구성된 프로젝트 실행 환경

## 진행 방식

- 모든 uv 명령은 **VS Code 터미널**에서 실행하세요.
- 명령을 실행하기 전에 현재 폴더를 확인하세요.
- `uv add`, `uv remove`, `uv sync`는 `pyproject.toml`이 있는 프로젝트 폴더에서 실행합니다.
- 패키지 버전, 설치 개수, 로그 문구는 실습 시점과 운영체제에 따라 달라질 수 있습니다.
- 결과 화면의 정확한 숫자보다 파일과 환경이 요구사항대로 변경되었는지 확인하세요.

## 심화 1. `.venv` 삭제 후 `uv sync` 복원

### 문제 4-1. 가상환경 유실 상황 복구

### 사전 확인

- 현재 위치가 `student_data_project`인지 확인하세요.
- `pyproject.toml`과 `uv.lock`은 삭제하지 않습니다.
- 실행 중인 Python 또는 Jupyter 커널이 `.venv`를 사용 중이라면 종료하세요.

### 1단계: `.venv` 삭제

Windows PowerShell:

```powershell
Remove-Item -Recurse -Force .venv
```

Windows 명령 프롬프트:

```cmd
rmdir /s /q .venv
```

Git Bash/macOS/Linux:

```bash
rm -rf .venv
```

### 2단계: 삭제 확인

Windows PowerShell:

```powershell
Test-Path .venv
```

Git Bash/macOS/Linux:

```bash
ls -a
```

### 3단계: 환경 복원

```bash
uv sync
```

### 4단계: 실행 검증

```bash
uv run python -c "import pandas; import matplotlib; print('환경 복원 성공')"
```

### 예상 핵심 흐름

```text
Creating virtual environment ...
Resolved ...
Installed ...
환경 복원 성공
```

## 자주 발생하는 오류 점검

### 1. `No pyproject.toml found`

uv 설치 또는 PATH 설정을 확인하세요.

### 4. `uv remove` 대상이 프로젝트 의존성에 없음

패키지명이 `pyproject.toml`에 직접 의존성으로 등록되어 있는지 확인하세요.

### 5. `.venv` 삭제 실패

가상환경을 사용하는 터미널, Python 프로세스, Jupyter 커널을 종료하세요.

현재 위치와 `pyproject.toml`, `uv.lock`, 실행 명령이 `uv run`인지 확인하세요.
