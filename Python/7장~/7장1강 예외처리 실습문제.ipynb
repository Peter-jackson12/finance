## 필수 2. 다중 예외와 `else`, `finally`

### 문제 2-1. 메뉴 인덱스 조회 프로그램

강의자료의 `IndexError`를 정확하게 확인하기 위해 사용자가 리스트 인덱스 `0~2`를 직접 입력하도록 구성합니다.

### 데이터

```python
menus = ["김치찌개", "비빔밥", "돈가스"]
```

### 요구사항
1. 위 리스트를 작성하세요.
2. 사용자에게 `0~2` 사이의 인덱스를 입력받으세요.
3. 입력값을 `int()`로 변환하세요.
4. 입력한 인덱스로 메뉴를 조회하세요.
5. 숫자가 아니면 `ValueError`를 처리하세요.
6. 존재하지 않는 인덱스면 `IndexError`를 처리하세요.
7. 예외가 없을 때만 `else`에서 결과를 출력하세요.
8. 성공·실패와 관계없이 `finally`에서 종료 문구를 출력하세요.

### 출력 문구

```text
메뉴 인덱스는 숫자로 입력해야 합니다.
```

```text
존재하지 않는 위치의 메뉴입니다.
```

```text
선택한 메뉴: 비빔밥
```

```text
메뉴 조회를 종료합니다.
```

````python
# 메뉴 조회와 다중 예외 처리를 작성하세요.
menus = ['김치찌개', '비빔밥', '돈가스']
try:
    menu_index = int(input('조회할 메뉴 인덱스 (0~2)입력하세요: '))
    selected_menu = menus[menu_index]
except ValueError:
    print('메뉴 인덱스는 숫자로 입력해야 합니다.')
except IndexError:
    print("존재하지 않는 위치의 메뉴입니다.")
else:
    print(f"선택한 메뉴: {selected_menu}")
finally:
    print('메뉴 조회를 종료합니다.')
````

### 테스트 값
- 정상 입력: `0`, `1`, `2`
- 숫자가 아닌 입력: `둘`
- 범위를 벗어난 입력: `3`, `5`

> 파이썬에서 `-1`은 마지막 요소를 의미하므로 `-1`은 `IndexError`가 아닙니다. 이번 문제에서는 양수 범위 초과 값으로 `IndexError`를 확인하세요.

## 심화 1. `raise`를 활용한 주문 수량 검증

### 문제 3-1. 주문 수량 검증 함수

### 요구사항
1. `validate_order_quantity(quantity)` 함수를 정의하세요.
2. 수량이 0 이하이면 다음 예외를 발생시키세요.

```python
raise ValueError("주문 수량은 1개 이상이어야 합니다.")
```

3. 정상 수량이면 다음 문자열을 반환하세요.

```text
주문 수량이 3개로 설정되었습니다.
```

4. 함수 밖에서 `try-except`로 호출하세요.
5. `except ValueError as error`로 예외 객체를 받으세요.
6. 오류 발생 시 다음 형식으로 출력하세요.

```text
주문 오류: 주문 수량은 1개 이상이어야 합니다.
```

7. `3`, `0`, `-2`를 각각 테스트하세요.

````python
# validate_order_quantity 함수를 작성하세요.
def validate_order_quantity(quantity):
    if quantity <= 0:
        raise ValueError("주문 수량은 1개 이상이어야 합니다.")
    return f"주문 수량이 {quantity}개로 설정되었습니다."

for quantity in [3, 0, -2]:
    try:
        print(validate_order_quantity(quantity))
    except ValueError as error:
        print(f"주문 오류: {error}")
````

````python
# 정상 수량과 잘못된 수량을 테스트하세요.

````

### 확인 사항
- `raise`는 문법상 가능한 값이라도 업무 규칙에 맞지 않을 때 의도적으로 예외를 발생시킵니다.
- 함수에서 발생한 예외는 호출한 코드의 `except`로 전달할 수 있습니다.

## 심화 2. 입력 변환과 업무 규칙 통합

### 문제 4-1. 주문 수량 입력 프로그램

### 요구사항
1. `input()`으로 주문 수량을 입력받으세요.
2. 입력값을 `int()`로 변환하세요.
3. `validate_order_quantity()`를 호출하세요.
4. 숫자가 아닌 입력과 0 이하 입력을 `ValueError`로 처리하세요.
5. 예외 객체를 사용하여 다음 형식으로 출력하세요.

```text
주문 처리 실패: 오류 메시지
```

6. 예외가 없으면 `else`에서 함수 반환값을 출력하세요.
7. `finally`에서 다음 문구를 출력하세요.

```text
주문 수량 검증을 종료합니다.
```

````python
# 입력 변환과 주문 수량 검증을 연결하세요.
try:
    quantity = int(input("주문 수량을 입력하세요."))
    message = validate_order_quantity(quantity)
except ValueError as error:
    print(f"주문 처리 실패: {error}")
else:
    print(message)
finally:
    print("주문 수량 검증 종료합니다.")
````

### 테스트 값
- 정상 입력: `5`
- 숫자가 아닌 입력: `다섯`
- 업무 규칙 위반: `0`, `-3`
