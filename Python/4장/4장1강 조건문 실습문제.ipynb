## 필수 1. 단일 `if`로 결제 완료 상태 확인하기

### 문제 1-1. 결제 완료 주문 알림

조건이 참일 때만 특정 코드가 실행되는 단일 `if` 구조를 작성합니다.

### 요구사항

1. `is_payment_complete`에 `True`를 저장하세요.
2. 값이 참일 때만 `"결제 완료 주문입니다."`를 출력하는 단일 `if` 문을 작성하세요.
3. 코드 블록 바깥에서 `"주문 상태 확인이 끝났습니다."`를 항상 출력하세요.
4. 첫 실행 후 값을 `False`로 변경하여 첫 번째 문구가 출력되지 않는지 확인하세요.

### 예상 결과

`True`일 때:

```text
결제 완료 주문입니다.
주문 상태 확인이 끝났습니다.
```

`False`일 때:

```text
주문 상태 확인이 끝났습니다.
```

````python
# 첫 번째 테스트: is_payment_complete = True
is_payment_complete = True

if is_payment_complete:
    print("결제 완료 주문입니다.")
print("주문 상태 확인이 끝났습니다.")
````

````python
# 두 번째 테스트: is_payment_complete = False
is_payment_complete = False

if is_payment_complete:
    print("결제 완료 주문입니다.")
print("주문 상태 확인이 끝났습니다.")
````

### 확인 사항

- 조건이 거짓이면 `if` 블록만 건너뛰는지 확인하세요.
- 들여쓰기되지 않은 문장은 조건과 관계없이 실행되는지 확인하세요.

## 필수 2. `if-else`로 재고 상태 판별하기

### 문제 2-1. 재고 부족 경고

재고가 기준 이하인지 판별하여 두 가지 안내 중 하나를 출력합니다.

### 요구사항

1. `stock_quantity`에 `4`를 저장하세요.
2. 재고가 `5` 이하이면 `"재고가 부족합니다. 발주를 확인하세요."`를 출력하세요.
3. 그렇지 않으면 `"재고가 충분합니다."`를 출력하세요.
4. 값을 `4`와 `12`로 각각 테스트하세요.

### 예상 결과

- `4`: 재고 부족 안내
- `12`: 재고 충분 안내

````python
# 첫 번째 테스트: stock_quantity = 4
stock_quantity = 4

if stock_quantity <= 5:
    print("재고가 부족합니다. 발주를 확인하세요.")
else:
    print("재고가 충분합니다.")
````

````python
# 두 번째 테스트: stock_quantity = 12
stock_quantity = 12

if stock_quantity <= 5:
    print("재고가 부족합니다. 발주를 확인하세요.")
else:
    print("재고가 충분합니다.")
````

## 필수 3. `if-elif-else`로 수하물 유형 분류하기

### 문제 3-1. 무게별 서비스 분류

### 요구사항

1. `weight`에 `20`을 저장하세요.
2. `weight > 23`이면 다음 문구를 출력하세요.

```text
대형 수하물 대상입니다. 추가 요금이 부과됩니다.
```

3. 첫 번째 조건이 거짓이고 `weight > 15`이면 다음 문구를 출력하세요.

```text
일반 위탁 수하물 대상입니다.
```

4. 위 조건이 모두 거짓이면 다음 문구를 출력하세요.

```text
무료 수하물 대상입니다.
```

5. `weight`를 `20`, `25`, `10`으로 바꾸어 세 분기를 모두 확인하세요.

````python
# 첫 번째 테스트: weight = 20
weight = 20

if weight > 23:
    print("대형 수하물 대상입니다. 추가 요금이 부과됩니다.")
elif weight > 15:
    print("일반 위탁 수하물 대상입니다.")
else:
    print("무료 수하물 대상입니다.")
````

````python
# 두 번째 테스트: weight = 25
weight = 25

if weight > 23:
    print("대형 수하물 대상입니다. 추가 요금이 부과됩니다.")
elif weight > 15:
    print("일반 위탁 수하물 대상입니다.")
else:
    print("무료 수하물 대상입니다.")

````

````python
# 세 번째 테스트: weight = 10
weight = 10

if weight > 23:
    print("대형 수하물 대상입니다. 추가 요금이 부과됩니다.")
elif weight > 15:
    print("일반 위탁 수하물 대상입니다.")
else:
    print("무료 수하물 대상입니다.")

````

## 심화 1. 중첩 조건문과 `and` 조건 비교하기

### 문제 4-1. 관리자 대시보드 접근 검사

로그인 여부와 관리자 권한을 모두 만족해야 접근할 수 있습니다.

### 요구사항

1. `is_logged_in = True`, `has_permission = True`를 저장하세요.
2. 먼저 `if` 문 안에 또 다른 `if` 문을 작성하는 중첩 조건문으로 접근 여부를 확인하세요.
3. 같은 조건을 `and`로 결합한 단일 `if-else` 문으로 다시 작성하세요.
4. 접근 가능할 때는 `"관리자 대시보드에 접근할 수 있습니다."`를 출력하세요.
5. 접근할 수 없을 때는 `"접근 권한이 없습니다."`를 출력하세요.
6. 다음 값도 테스트하세요.
   - `is_logged_in = False`, `has_permission = True`
   - `is_logged_in = True`, `has_permission = False`

````python
# 1단계: 중첩 조건문으로 작성하세요.
is_logged_in = True
has_permission = True

if is_logged_in:
    if has_permission:
        print("관리자 대시보드에 접근할 수 있습니다.")
    else:
        print("접근 권한이 없습니다.")
````

````python
# 2단계: and를 사용한 단일 조건문으로 작성하세요.
is_logged_in = True
has_permission = True

if is_logged_in and has_permission:
    print("관리자 대시보드에 접근할 수 있습니다.")
else:
    print("접근 권한이 없습니다.")
````

````python
# 추가 테스트 1: 로그인하지 않은 경우
is_logged_in = False
has_permission = True

if is_logged_in and has_permission:
    print("관리자 대시보드에 접근할 수 있습니다.")
else:
    print("접근 권한이 없습니다.")
````

````python
# 추가 테스트 2: 권한이 없는 경우
is_logged_in = True
has_permission = False

if is_logged_in and has_permission:
    print("관리자 대시보드에 접근할 수 있습니다.")
else:
    print("접근 권한이 없습니다.")
````

### 확인 사항

- 중첩 구조와 `and` 구조가 같은 판정 결과를 만드는지 확인하세요.
- `and`에서는 두 조건이 모두 참일 때만 전체 조건이 참입니다.

## 심화 2. `or`로 무료 배송 대상 판별하기

### 문제 5-1. 회원 등급 또는 구매 금액 기준 혜택

VIP 고객이거나 구매 금액이 100,000원 이상이면 무료 배송을 제공합니다.

### 요구사항

1. `is_vip = False`, `purchase_amount = 120000`을 저장하세요.
2. `or`를 사용하여 두 조건 중 하나라도 참이면 `"무료 배송 대상입니다."`를 출력하세요.
3. 두 조건이 모두 거짓이면 `"배송비가 부과됩니다."`를 출력하세요.
4. 아래 값을 각각 테스트하세요.
   - `is_vip = False`, `purchase_amount = 120000`
   - `is_vip = True`, `purchase_amount = 30000`
   - `is_vip = False`, `purchase_amount = 30000`

````python
# 첫 번째 테스트: 구매 금액 조건 충족
is_vip = False
purchase_amount = 120000

if is_vip or purchase_amount >= 100000:
    print("무료 배송 대상입니다.")
else:
    print("배송비가 부과됩니다.")
````

````python
# 두 번째 테스트: VIP 조건 충족
is_vip = True
purchase_amount = 30000

if is_vip or purchase_amount >= 100000:
    print("무료 배송 대상입니다.")
else:
    print("배송비가 부과됩니다.")
````

````python
# 세 번째 테스트: 두 조건 모두 미충족
is_vip = False
purchase_amount = 30000

if is_vip or purchase_amount >= 100000:
    print("무료 배송 대상입니다.")
else:
    print("배송비가 부과됩니다.")
````
