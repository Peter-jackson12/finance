```python
def solution(study_times, min_time):
    count = 0

    for time in study_times:
        if time >= min_time:
            count += 1

    return count
```

```python
solution([10, 20, 25, 15], 30)
```

```python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department
        pass

employee = Employee("김민수", "데이터팀")

print(employee.name)
print(employee.department)
```

```python
class Member:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        print(f"안녕하세요. 저는 {self.name}입니다.")

member = Member('지수')
member.introduce()
```

```python
class Character:
    def __init__(self, name):
        self.name = name
        self.hp = 50

    def heal(self):
        self.hp += 20
        print(f"{self.name}의 체력: {self.hp}")

character = Character("마법사")

print(character.hp)
character.heal()
```

```python
# 연습문제 5
class Product:
    def __init__(self, name, stock):
        self.name = name
        self.stock = stock

    def sell(self):
        self.stock -= 1
        print(f'{self.name} 남은 재고: {self.stock}개')
        pass

product = Product("키보드", 5)
# product 인스턴스 안에 있는 변수 name = 키보드, product 인스턴스 안에 있는 stock은 5가 됨
product.sell()
# product 인스턴스 안에 있는 stock 값이 -1, 상품 이름 남은 재고 : 몇 개
```
